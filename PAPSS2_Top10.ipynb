```python

# ============================================================
# STEP 7: MATCH DOCKING RECORDS WITH SDF STRUCTURES
# ============================================================

import io
import os
import re
import pandas as pd
from rdkit import Chem

# Reload docking Excel
excel_file = list(uploaded_docking.keys())[0]
df = pd.read_excel(excel_file)

# Function to standardize PubChem CID
def clean_cid(value):

    if pd.isna(value):
        return None

    try:
        return str(int(float(value)))
    except:
        return str(value).strip()


# Function to standardize library names
def clean_parent(value):

    if pd.isna(value):
        return None

    value = str(value).strip()

    value = re.split(
        r"_derivatives",
        value,
        flags=re.IGNORECASE
    )[0]

    return value.lower()


# ------------------------------------------------------------
# PART A: READ ALL SDF STRUCTURES
# ------------------------------------------------------------

sdf_records = []

for filename, content in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(content),
        removeHs=False,
        sanitize=True
    )

    for mol in supplier:

        if mol is None:
            continue

        if not mol.HasProp("PubChem_CID"):
            continue

        cid = clean_cid(mol.GetProp("PubChem_CID"))

        parent = clean_parent(
            mol.GetProp("Parent_Ligand")
            if mol.HasProp("Parent_Ligand")
            else None
        )

        record = {
            "CID": cid,
            "Parent": parent,
            "Title": mol.GetProp("PubChem_Title")
                     if mol.HasProp("PubChem_Title") else "",
            "IUPAC": mol.GetProp("IUPAC_Name")
                     if mol.HasProp("IUPAC_Name") else "",
            "Formula": mol.GetProp("Molecular_Formula")
                       if mol.HasProp("Molecular_Formula") else "",
            "SDF_File": filename,
            "Molecule": mol
        }

        sdf_records.append(record)


print("TOTAL SDF STRUCTURES READ:", len(sdf_records))


# ------------------------------------------------------------
# PART B: CREATE SDF LOOKUP
# ------------------------------------------------------------

sdf_lookup = {}

for record in sdf_records:

    key = (record["CID"], record["Parent"])

    sdf_lookup.setdefault(key, []).append(record)


# Also create CID-only lookup for investigating unmatched rows
cid_lookup = {}

for record in sdf_records:

    cid_lookup.setdefault(record["CID"], []).append(record)


# ------------------------------------------------------------
# PART C: MATCH DOCKING DATA
# ------------------------------------------------------------

df["Clean_CID"] = df["PubChem CID"].apply(clean_cid)

df["Clean_Parent"] = df["Library"].apply(clean_parent)

match_results = []

for _, row in df.iterrows():

    key = (
        row["Clean_CID"],
        row["Clean_Parent"]
    )

    candidates = sdf_lookup.get(key, [])

    if len(candidates) == 1:

        status = "MATCHED"
        selected = candidates[0]

    elif len(candidates) > 1:

        status = "AMBIGUOUS_MULTIPLE_MATCHES"
        selected = None

    else:

        cid_candidates = cid_lookup.get(
            row["Clean_CID"], []
        )

        if len(cid_candidates) == 1:

            status = "CID_MATCH_PARENT_MISMATCH"
            selected = cid_candidates[0]

        elif len(cid_candidates) > 1:

            status = "AMBIGUOUS_CID"
            selected = None

        else:

            status = "NO_SDF_MATCH"
            selected = None

    match_results.append({

        "IX": row["IX"],
        "PubChem_CID": row["Clean_CID"],
        "Library": row["Library"],
        "Edock": row["Edock"],
        "Match_Status": status,

        "Matched_Title":
            selected["Title"] if selected else "",

        "Matched_Formula":
            selected["Formula"] if selected else "",

        "Matched_SDF":
            selected["SDF_File"] if selected else ""
    })


# ------------------------------------------------------------
# PART D: MATCHING REPORT
# ------------------------------------------------------------

match_df = pd.DataFrame(match_results)

print("\nMATCHING SUMMARY")

print("\nTotal docking records:", len(match_df))

print("\nMATCH STATUS:")
print(match_df["Match_Status"].value_counts())

print("\nMATCHED RECORDS:")
display(
    match_df[
        match_df["Match_Status"] == "MATCHED"
    ].head(20)
)

print("\nUNMATCHED OR AMBIGUOUS RECORDS:")
display(
    match_df[
        match_df["Match_Status"] != "MATCHED"
    ]
)

print("\nUNIQUE MATCHED COMPOUNDS:",
      match_df[
          match_df["Match_Status"] == "MATCHED"
      ]["PubChem_CID"].nunique()
)

# Save audit
match_df.to_csv(
    "STEP7_SDF_Docking_Matching_Audit.csv",
    index=False
)

print("\nMatching audit saved successfully.")

```

```python
# ============================================================
# STEP 2: LIGAND EFFICIENCY AND STRUCTURAL DESCRIPTORS
# CORRECTED VERSION
# ============================================================

import pandas as pd
import numpy as np
import io

from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, Crippen

# ------------------------------------------------------------
# 1. LOAD DOCKING RESULTS
# ------------------------------------------------------------

docking_file = "PAPSS2_8I1O_Top10_All_24_Libraries_3D_docked.xlsx"

df = pd.read_excel(docking_file)

# Standardize CID format
df["PubChem CID"] = pd.to_numeric(
    df["PubChem CID"], errors="coerce"
).astype("Int64")

df["Edock"] = pd.to_numeric(
    df["Edock"], errors="coerce"
)

# ------------------------------------------------------------
# 2. CREATE SDF STRUCTURE LOOKUP
# ------------------------------------------------------------

sdf_lookup = {}

invalid_sdf = []

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=False,
        removeHs=False
    )

    for mol in supplier:

        if mol is None:
            invalid_sdf.append(filename)
            continue

        if not mol.HasProp("PubChem_CID"):
            continue

        cid = mol.GetProp("PubChem_CID").strip()

        if cid not in sdf_lookup:
            sdf_lookup[cid] = mol

# ------------------------------------------------------------
# 3. CALCULATE STRUCTURAL DESCRIPTORS
# ------------------------------------------------------------

results = []

for _, row in df.iterrows():

    cid = str(int(row["PubChem CID"]))

    mol = sdf_lookup.get(cid)

    if mol is None:

        results.append({
            "PubChem CID": int(cid),
            "Structure Status": "Not found"
        })

        continue

    try:

        # Make a copy before sanitization
        mol_clean = Chem.Mol(mol)

        Chem.SanitizeMol(mol_clean)

        mol_clean = Chem.RemoveHs(mol_clean)

        heavy_atoms = mol_clean.GetNumHeavyAtoms()

        # Supporting ligand-efficiency measure
        ligand_efficiency = (
            abs(row["Edock"]) / heavy_atoms
            if heavy_atoms > 0 else np.nan
        )

        results.append({

            "PubChem CID": int(cid),

            "Structure Status": "Valid",

            "Heavy Atom Count": heavy_atoms,

            "Molecular Weight": Descriptors.MolWt(mol_clean),

            "HBD": Lipinski.NumHDonors(mol_clean),

            "HBA": Lipinski.NumHAcceptors(mol_clean),

            "cLogP": Crippen.MolLogP(mol_clean),

            "Ligand Efficiency Screening": ligand_efficiency

        })

    except Exception:

        results.append({

            "PubChem CID": int(cid),

            "Structure Status": "Invalid structure"

        })

# ------------------------------------------------------------
# 4. MERGE RESULTS
# ------------------------------------------------------------

descriptor_df = pd.DataFrame(results)

descriptor_df["PubChem CID"] = pd.to_numeric(
    descriptor_df["PubChem CID"], errors="coerce"
).astype("Int64")

analysis_df = df.merge(
    descriptor_df,
    on="PubChem CID",
    how="left"
)

# ------------------------------------------------------------
# 5. RANK BY EDOCK
# ------------------------------------------------------------

analysis_df = analysis_df.sort_values(
    "Edock",
    ascending=True
).reset_index(drop=True)

analysis_df.insert(
    0,
    "Edock Rank",
    range(1, len(analysis_df) + 1)
)

# ------------------------------------------------------------
# 6. SAVE RESULTS
# ------------------------------------------------------------

analysis_df.to_csv(
    "PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv",
    index=False
)

# ------------------------------------------------------------
# 7. DISPLAY SUMMARY
# ------------------------------------------------------------

print("TOTAL DOCKED COMPOUNDS:", len(analysis_df))

print("\nSTRUCTURE STATUS:")
print(analysis_df["Structure Status"].value_counts(dropna=False))

print("\nTOP 20 COMPOUNDS:\n")

display(
    analysis_df[
        [
            "Edock Rank",
            "PubChem CID",
            "Library",
            "Edock",
            "Heavy Atom Count",
            "Ligand Efficiency Screening",
            "Molecular Weight",
            "HBD",
            "HBA",
            "cLogP",
            "Structure Status"
        ]
    ].head(20)
)

print("\nSAVED:")
print("PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv")
```

```python
# ============================================================
# STEP 3: CHEMICAL DIVERSITY ANALYSIS OF TOP 30
# ============================================================

import pandas as pd
import io

from rdkit import Chem, DataStructs
from rdkit.Chem import rdFingerprintGenerator

# Load the 236-compound analysis
analysis_df = pd.read_csv(
    "PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv"
)

# Select top 30 by Edock
top30 = analysis_df.sort_values(
    "Edock",
    ascending=True
).head(30).copy()

# Morgan fingerprint generator
generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

# Create CID-to-structure lookup
molecule_lookup = {}

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=True,
        removeHs=True
    )

    for mol in supplier:

        if mol is None:
            continue

        if mol.HasProp("PubChem_CID"):

            cid = mol.GetProp("PubChem_CID").strip()

            if cid not in molecule_lookup:
                molecule_lookup[cid] = mol

# Generate fingerprints
fingerprints = []
valid_rows = []

for _, row in top30.iterrows():

    cid = str(int(row["PubChem CID"]))

    mol = molecule_lookup.get(cid)

    if mol is not None:

        fingerprints.append(
            generator.GetFingerprint(mol)
        )

        valid_rows.append(row)

# Calculate pairwise similarities
similarity_results = []

for i in range(len(fingerprints)):

    for j in range(i + 1, len(fingerprints)):

        similarity = DataStructs.TanimotoSimilarity(
            fingerprints[i],
            fingerprints[j]
        )

        if similarity >= 0.70:

            similarity_results.append({

                "CID 1": int(valid_rows[i]["PubChem CID"]),

                "CID 2": int(valid_rows[j]["PubChem CID"]),

                "Edock 1": valid_rows[i]["Edock"],

                "Edock 2": valid_rows[j]["Edock"],

                "Tanimoto Similarity": similarity

            })

similarity_df = pd.DataFrame(similarity_results)

print("Top 30 compounds examined:", len(top30))
print("Valid fingerprints:", len(fingerprints))

print("\nHighly similar pairs (Tanimoto >= 0.70):")

if len(similarity_df) > 0:

    display(
        similarity_df.sort_values(
            "Tanimoto Similarity",
            ascending=False
        )
    )

else:

    print("No highly similar pairs detected.")

similarity_df.to_csv(
    "PAPSS2_TOP30_CHEMICAL_SIMILARITY.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP30_CHEMICAL_SIMILARITY.csv")
```

```python
# ============================================================
# STEP 4: IDENTIFY UNIQUE STRUCTURES AMONG TOP 30
# ============================================================

import pandas as pd
import io

from rdkit import Chem, DataStructs
from rdkit.Chem import rdFingerprintGenerator

# Load complete analysis
analysis_df = pd.read_csv(
    "PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv"
)

# Select top 30 by Edock
top30 = analysis_df.sort_values(
    "Edock",
    ascending=True
).head(30).copy()

# Morgan fingerprint generator
generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

# Retrieve structures
molecule_lookup = {}

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=True,
        removeHs=True
    )

    for mol in supplier:

        if mol is not None and mol.HasProp("PubChem_CID"):

            cid = mol.GetProp("PubChem_CID").strip()

            if cid not in molecule_lookup:
                molecule_lookup[cid] = mol

# Generate fingerprints
fingerprints = []

for _, row in top30.iterrows():

    cid = str(int(row["PubChem CID"]))

    mol = molecule_lookup[cid]

    fp = generator.GetFingerprint(mol)

    fingerprints.append(fp)

# Group compounds with identical fingerprints
groups = []
assigned = set()

for i in range(len(fingerprints)):

    if i in assigned:
        continue

    group = [i]

    for j in range(i + 1, len(fingerprints)):

        if j in assigned:
            continue

        similarity = DataStructs.TanimotoSimilarity(
            fingerprints[i],
            fingerprints[j]
        )

        if similarity == 1.0:
            group.append(j)
            assigned.add(j)

    assigned.add(i)
    groups.append(group)

# Create group report
group_results = []

for group_number, group in enumerate(groups, start=1):

    for index in group:

        row = top30.iloc[index]

        group_results.append({

            "Structure Group": group_number,

            "PubChem CID": int(row["PubChem CID"]),

            "Edock Rank": int(row["Edock Rank"]),

            "Edock": row["Edock"],

            "Ligand Efficiency Screening":
                row["Ligand Efficiency Screening"],

            "Library": row["Library"],

            "Group Representative":
                index == min(group)

        })

group_df = pd.DataFrame(group_results)

# Display results
print("TOTAL COMPOUNDS:", len(top30))
print("UNIQUE FINGERPRINT GROUPS:", len(groups))

print("\nSTRUCTURE GROUPING:\n")

display(
    group_df.sort_values(
        ["Structure Group", "Edock Rank"]
    )
)

# Save
group_df.to_csv(
    "PAPSS2_TOP30_UNIQUE_STRUCTURE_GROUPS.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP30_UNIQUE_STRUCTURE_GROUPS.csv")
```

```python
# ============================================================
# STEP 5: VERIFY EXACT STRUCTURAL IDENTITY
# ============================================================

import pandas as pd
import io

from rdkit import Chem

# Load top 30
analysis_df = pd.read_csv(
    "PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv"
)

top30 = analysis_df.sort_values(
    "Edock",
    ascending=True
).head(30)

# Create CID-to-canonical-SMILES lookup
smiles_lookup = {}

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=True,
        removeHs=True
    )

    for mol in supplier:

        if mol is None:
            continue

        if mol.HasProp("PubChem_CID"):

            cid = mol.GetProp("PubChem_CID").strip()

            if cid not in smiles_lookup:

                smiles_lookup[cid] = Chem.MolToSmiles(
                    mol,
                    canonical=True,
                    isomericSmiles=True
                )

# Compare structures within each fingerprint group
group_df = pd.read_csv(
    "PAPSS2_TOP30_UNIQUE_STRUCTURE_GROUPS.csv"
)

results = []

for group_number, group in group_df.groupby("Structure Group"):

    group_smiles = []

    for _, row in group.iterrows():

        cid = str(int(row["PubChem CID"]))

        group_smiles.append({
            "CID": cid,
            "Edock": row["Edock"],
            "SMILES": smiles_lookup.get(cid, "NOT FOUND")
        })

    unique_smiles = set(
        item["SMILES"] for item in group_smiles
    )

    results.append({

        "Structure Group": group_number,

        "Number of Compounds": len(group_smiles),

        "Unique Canonical SMILES": len(unique_smiles),

        "Exact Structures Identical":
            len(unique_smiles) == 1

    })

identity_df = pd.DataFrame(results)

print("EXACT STRUCTURAL IDENTITY AUDIT\n")

display(identity_df)

identity_df.to_csv(
    "PAPSS2_TOP30_EXACT_STRUCTURE_AUDIT.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP30_EXACT_STRUCTURE_AUDIT.csv")
```

```python
# ============================================================
# STEP 8: EDOCK OUTLIER INVESTIGATION
# ============================================================

import pandas as pd
import numpy as np

# Load all 236 docking results
df = pd.read_excel(
    "PAPSS2_8I1O_Top10_All_24_Libraries_3D_docked.xlsx"
)

# Convert scores to numeric
score_columns = ["Edock", "Egv", "Egb", "Ege", "Egs"]

for col in score_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Calculate IQR limits for Edock
Q1 = df["Edock"].quantile(0.25)
Q3 = df["Edock"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[
    (df["Edock"] < lower_limit) |
    (df["Edock"] > upper_limit)
].copy()

# Rank all compounds
df = df.sort_values(
    "Edock",
    ascending=True
).reset_index(drop=True)

df.insert(0, "Edock Rank", range(1, len(df) + 1))

print("EDOCK OUTLIER ANALYSIS")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

print("\nNumber of Edock outliers:", len(outliers))

print("\nOUTLIER COMPOUNDS:\n")

display(
    outliers[
        [
            "PubChem CID",
            "Molecule Name",
            "Library",
            "Edock",
            "Egv",
            "Egb",
            "Ege",
            "Egs"
        ]
    ].sort_values("Edock")
)

print("\nTOP 10 EDOCK SCORES WITH OUTLIER FLAGS:\n")

df["Edock_Outlier"] = (
    (df["Edock"] < lower_limit) |
    (df["Edock"] > upper_limit)
)

display(
    df[
        [
            "Edock Rank",
            "PubChem CID",
            "Edock",
            "Egv",
            "Egb",
            "Ege",
            "Egs",
            "Edock_Outlier"
        ]
    ].head(15)
)

# Save
outliers.to_csv(
    "PAPSS2_EDOCK_OUTLIERS.csv",
    index=False
)

print("\nSaved: PAPSS2_EDOCK_OUTLIERS.csv")
```

```python
# ============================================================
# STEP 10: FIRST-STAGE VS SECOND-STAGE EDOCK COMPARISON
# ============================================================

import pandas as pd
from scipy.stats import spearmanr, pearsonr

file_name = "PAPSS2_8I1O_Top10_All_24_Libraries_3D_docked.xlsx"
df = pd.read_excel(file_name)

df["PubChem CID"] = pd.to_numeric(df["PubChem CID"], errors="coerce")
df["Edock"] = pd.to_numeric(df["Edock"], errors="coerce")
df["T tmp chem Edock"] = pd.to_numeric(
    df["T tmp chem Edock"], errors="coerce"
)

comparison = df[
    ["PubChem CID", "Molecule Name", "Library",
     "T tmp chem Edock", "Edock"]
].dropna(subset=["T tmp chem Edock", "Edock"]).copy()

comparison["Score Change"] = (
    comparison["Edock"] - comparison["T tmp chem Edock"]
)

comparison["First Stage Rank"] = (
    comparison["T tmp chem Edock"]
    .rank(method="min", ascending=True)
    .astype(int)
)

comparison["Second Stage Rank"] = (
    comparison["Edock"]
    .rank(method="min", ascending=True)
    .astype(int)
)

comparison["Rank Shift"] = (
    comparison["First Stage Rank"] -
    comparison["Second Stage Rank"]
)

print("NUMBER OF COMPOUNDS:", len(comparison))

print("\nCORRELATION BETWEEN FIRST- AND SECOND-STAGE SCORES")
print("Pearson:", pearsonr(
    comparison["T tmp chem Edock"], comparison["Edock"]
))
print("Spearman:", spearmanr(
    comparison["T tmp chem Edock"], comparison["Edock"]
))

print("\nLARGEST ABSOLUTE SCORE CHANGES")
display(
    comparison.assign(
        Absolute_Change=comparison["Score Change"].abs()
    ).sort_values("Absolute_Change", ascending=False).head(15)
)

print("\nTOP 15 BY SECOND-STAGE EDOCK")
display(
    comparison.sort_values("Edock", ascending=True).head(15)
)

comparison.to_csv(
    "PAPSS2_FIRST_VS_SECOND_STAGE_EDOCK_COMPARISON.csv",
    index=False
)

print("\nSaved: PAPSS2_FIRST_VS_SECOND_STAGE_EDOCK_COMPARISON.csv")
```

```python
# ============================================================
# STEP 11: CONSOLIDATED TOP 15 CANDIDATE REVIEW
# ============================================================

import pandas as pd

docking_file = "PAPSS2_8I1O_Top10_All_24_Libraries_3D_docked.xlsx"
descriptor_file = "PAPSS2_ALL_236_LIGAND_EFFICIENCY_ANALYSIS.csv"

docking = pd.read_excel(docking_file)
descriptors = pd.read_csv(descriptor_file)

docking["PubChem CID"] = pd.to_numeric(
    docking["PubChem CID"], errors="coerce"
)

docking["Edock"] = pd.to_numeric(
    docking["Edock"], errors="coerce"
)

docking["T tmp chem Edock"] = pd.to_numeric(
    docking["T tmp chem Edock"], errors="coerce"
)

# Rank using second-stage Edock
docking["Edock Rank"] = docking["Edock"].rank(
    method="min", ascending=True
).astype(int)

# Keep top 15
top15 = docking.nsmallest(15, "Edock").copy()

# Merge structural descriptors
review = top15.merge(
    descriptors,
    on="PubChem CID",
    how="left",
    suffixes=("", "_descriptor")
)

columns_to_show = [
    "Edock Rank",
    "PubChem CID",
    "Library",
    "T tmp chem Edock",
    "Edock",
    "Egv",
    "Egb",
    "Ege",
    "Egs",
    "Heavy Atom Count",
    "Molecular Weight",
    "HBD",
    "HBA",
    "cLogP",
    "Ligand Efficiency"
]

available_columns = [
    col for col in columns_to_show if col in review.columns
]

review = review[available_columns].sort_values("Edock Rank")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

display(review)

review.to_csv(
    "PAPSS2_TOP15_FINAL_CANDIDATE_REVIEW.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP15_FINAL_CANDIDATE_REVIEW.csv")
```

```python
# ============================================================
# STEP 12: EXACT STRUCTURE DUPLICATE AUDIT OF TOP 15
# ============================================================

from rdkit import Chem
from rdkit.Chem import rdMolDescriptors
import pandas as pd

# Use the previously matched source SDF structures
# from the existing uploaded_sdf dictionary.

top15_cids = [
    49849336, 153531369, 153531370, 175018237,
    150310321, 58965179, 21260039, 23187967,
    14157051, 68870353, 91337656, 14209308,
    129842494, 413359, 642841
]

# Search all uploaded SDF files
structure_records = []

for filename, sdf_bytes in uploaded_sdf.items():
    import io

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=True,
        removeHs=True
    )

    for mol in supplier:
        if mol is None:
            continue

        cid = mol.GetProp("PUBCHEM_COMPOUND_CID") if mol.HasProp(
            "PUBCHEM_COMPOUND_CID"
        ) else None

        if cid is None:
            continue

        try:
            cid = int(cid)
        except:
            continue

        if cid not in top15_cids:
            continue

        smiles = Chem.MolToSmiles(
            mol,
            canonical=True,
            isomericSmiles=True
        )

        structure_records.append({
            "PubChem CID": cid,
            "Source SDF": filename,
            "Canonical SMILES": smiles
        })

structures = pd.DataFrame(structure_records)

# One structure record per CID
structures = structures.drop_duplicates(
    subset=["PubChem CID", "Canonical SMILES"]
)

print("TOP 15 CIDs:", len(top15_cids))
print("Matched structure records:", len(structures))
print("Unique CIDs matched:", structures["PubChem CID"].nunique())

print("\nEXACT STRUCTURE DUPLICATES")
duplicate_groups = (
    structures.groupby("Canonical SMILES")["PubChem CID"]
    .agg(list)
)

duplicates = duplicate_groups[
    duplicate_groups.apply(lambda x: len(set(x)) > 1)
]

if len(duplicates) == 0:
    print("No exact structural duplicates found.")
else:
    for smiles, cids in duplicates.items():
        print("CIDs:", sorted(set(cids)))

print("\nSTRUCTURE MATCHING STATUS")
display(
    structures[["PubChem CID", "Source SDF"]]
    .sort_values("PubChem CID")
)

structures.to_csv(
    "PAPSS2_TOP15_EXACT_STRUCTURE_AUDIT.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP15_EXACT_STRUCTURE_AUDIT.csv")
```

```python
# ============================================================
# STEP 12B: EXACT STRUCTURE DUPLICATE AUDIT OF TOP 15
# Correct SDF property: PubChem_CID
# ============================================================

from rdkit import Chem
import pandas as pd
import io

top15_cids = [
    49849336, 153531369, 153531370, 175018237,
    150310321, 58965179, 21260039, 23187967,
    14157051, 68870353, 91337656, 14209308,
    129842494, 413359, 642841
]

structure_records = []
invalid_candidates = []

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=False,
        removeHs=True
    )

    for mol in supplier:

        if mol is None:
            continue

        if not mol.HasProp("PubChem_CID"):
            continue

        try:
            cid = int(mol.GetProp("PubChem_CID"))
        except:
            continue

        if cid not in top15_cids:
            continue

        mol_copy = Chem.Mol(mol)

        try:
            Chem.SanitizeMol(mol_copy)

            smiles = Chem.MolToSmiles(
                mol_copy,
                canonical=True,
                isomericSmiles=True
            )

            structure_records.append({
                "PubChem CID": cid,
                "Source SDF": filename,
                "Canonical SMILES": smiles,
                "Structure Status": "Valid"
            })

        except Exception as e:
            invalid_candidates.append({
                "PubChem CID": cid,
                "Source SDF": filename,
                "Error": str(e)
            })

structures = pd.DataFrame(
    structure_records,
    columns=[
        "PubChem CID",
        "Source SDF",
        "Canonical SMILES",
        "Structure Status"
    ]
)

print("TOP 15 CIDs:", len(top15_cids))
print("Matched valid structures:", structures["PubChem CID"].nunique())
print("Invalid candidate structures:", len(invalid_candidates))
print("Missing CIDs:", sorted(
    set(top15_cids) - set(structures["PubChem CID"])
))

print("\nEXACT STRUCTURE DUPLICATE AUDIT")

duplicates = (
    structures.groupby("Canonical SMILES")["PubChem CID"]
    .agg(lambda x: sorted(set(x)))
)

duplicates = duplicates[duplicates.apply(len) > 1]

if duplicates.empty:
    print("No exact structural duplicates found.")
else:
    for smiles, cids in duplicates.items():
        print("Duplicate CIDs:", cids)

print("\nSTRUCTURE MATCHING STATUS")
display(
    structures[
        ["PubChem CID", "Source SDF", "Structure Status"]
    ].sort_values("PubChem CID")
)

if invalid_candidates:
    print("\nINVALID CANDIDATE STRUCTURES")
    display(pd.DataFrame(invalid_candidates))

structures.to_csv(
    "PAPSS2_TOP15_EXACT_STRUCTURE_AUDIT.csv",
    index=False
)

print("\nSaved: PAPSS2_TOP15_EXACT_STRUCTURE_AUDIT.csv")
```

```python
# ============================================================
# FINAL STEP: EXTRACT TOP 10 PAPSS2 COMPOUNDS FROM 236
# Ranking criterion: Second-stage ICM-Pro Edock
# Lower (more negative) Edock = better ranking
# ============================================================

import pandas as pd

# Input workbook
input_file = "PAPSS2_8I1O_Top10_All_24_Libraries_3D_docked.xlsx"

# Read docking results
df = pd.read_excel(input_file)

# Ensure numerical values
df["PubChem CID"] = pd.to_numeric(
    df["PubChem CID"], errors="coerce"
)

df["Edock"] = pd.to_numeric(
    df["Edock"], errors="coerce"
)

# Remove records without CID or second-stage Edock
df = df.dropna(
    subset=["PubChem CID", "Edock"]
).copy()

# Verify unique compounds
df = df.drop_duplicates(
    subset=["PubChem CID"]
)

# Rank by second-stage Edock
df = df.sort_values(
    by="Edock",
    ascending=True
).reset_index(drop=True)

df["Final Rank"] = range(1, len(df) + 1)

# Extract final 10
top10 = df.head(10).copy()

# Select relevant columns
columns = [
    "Final Rank",
    "PubChem CID",
    "Molecule Name",
    "Library",
    "Library Rank",
    "Edock",
    "Egv",
    "Egb",
    "Ege",
    "Egs",
    "T tmp chem Edock",
    "CONF",
    "RecConf"
]

columns = [col for col in columns if col in top10.columns]

top10 = top10[columns]

# Display results
print("TOTAL UNIQUE COMPOUNDS:", len(df))
print("FINAL SELECTED COMPOUNDS:", len(top10))

print("\nFINAL TOP 10 PAPSS2 COMPOUNDS")
display(top10)

# Save CSV
top10.to_csv(
    "PAPSS2_FINAL_TOP10_COMPOUNDS.csv",
    index=False
)

# Save Excel
top10.to_excel(
    "PAPSS2_FINAL_TOP10_COMPOUNDS.xlsx",
    index=False
)

print("\nFILES SAVED:")
print("1. PAPSS2_FINAL_TOP10_COMPOUNDS.csv")
print("2. PAPSS2_FINAL_TOP10_COMPOUNDS.xlsx")
```

```python
# ============================================================
# FINAL TOP 10 PAPSS2 COMPOUNDS — COMBINED SDF
# Target: PAPSS2 (PDB: 8I10)
# Source: 24 uploaded ligand SDF files
# ============================================================

from rdkit import Chem
import pandas as pd
import io
import os

# ------------------------------------------------------------
# 1. FINAL TOP 10 PUBCHEM CIDs
# ------------------------------------------------------------

top10_cids = [
    49849336,
    153531369,
    153531370,
    175018237,
    150310321,
    58965179,
    21260039,
    23187967,
    14157051,
    68870353
]

# ------------------------------------------------------------
# 2. OUTPUT SDF FILE
# ------------------------------------------------------------

output_file = "/content/PAPSS2_FINAL_TOP10_COMPOUNDS.sdf"

writer = Chem.SDWriter(output_file)

exported_cids = set()
export_records = []

# ------------------------------------------------------------
# 3. SEARCH ALL UPLOADED SOURCE SDF FILES
# ------------------------------------------------------------

for filename, sdf_bytes in uploaded_sdf.items():

    supplier = Chem.ForwardSDMolSupplier(
        io.BytesIO(sdf_bytes),
        sanitize=False,
        removeHs=False
    )

    for mol in supplier:

        if mol is None:
            continue

        if not mol.HasProp("PubChem_CID"):
            continue

        try:
            cid = int(mol.GetProp("PubChem_CID"))
        except:
            continue

        if cid not in top10_cids:
            continue

        # Avoid exporting duplicates
        if cid in exported_cids:
            continue

        mol_copy = Chem.Mol(mol)

        try:
            Chem.SanitizeMol(mol_copy)
        except Exception:
            print("INVALID STRUCTURE — CID:", cid)
            continue

        # Add final ranking information
        rank = top10_cids.index(cid) + 1

        mol_copy.SetProp("Final_Rank", str(rank))
        mol_copy.SetProp("PubChem_CID", str(cid))
        mol_copy.SetProp("Target", "PAPSS2")
        mol_copy.SetProp("Target_PDB", "8I10")
        mol_copy.SetProp("Selection_Criteria", "Second-stage ICM-Pro Edock")

        # Write molecule
        writer.write(mol_copy)

        exported_cids.add(cid)

        export_records.append({
            "Final Rank": rank,
            "PubChem CID": cid,
            "Source SDF": filename,
            "Structure Status": "Valid"
        })

writer.close()

# ------------------------------------------------------------
# 4. EXPORT AUDIT
# ------------------------------------------------------------

export_df = pd.DataFrame(export_records)

print("\n========== TOP 10 SDF EXPORT AUDIT ==========")

print("Expected compounds:", len(top10_cids))
print("Successfully exported:", len(exported_cids))
print("Missing CIDs:", sorted(set(top10_cids) - exported_cids))

display(export_df.sort_values("Final Rank"))

# ------------------------------------------------------------
# 5. VERIFY THE GENERATED SDF
# ------------------------------------------------------------

check_supplier = Chem.SDMolSupplier(
    output_file,
    sanitize=True,
    removeHs=False
)

verified_molecules = [
    mol for mol in check_supplier if mol is not None
]

verified_cids = [
    int(mol.GetProp("PubChem_CID"))
    for mol in verified_molecules
    if mol.HasProp("PubChem_CID")
]

print("\n========== FINAL FILE VERIFICATION ==========")
print("SDF molecules readable:", len(verified_molecules))
print("Unique CIDs in SDF:", len(set(verified_cids)))
print("File size (KB):", round(os.path.getsize(output_file) / 1024, 2))

if (
    len(exported_cids) == 10
    and len(verified_molecules) == 10
    and set(verified_cids) == set(top10_cids)
):
    print("\nSUCCESS: All 10 compounds exported and verified.")
    print("OUTPUT:", output_file)
else:
    print("\nWARNING: Export verification failed. Check missing or invalid CIDs.")

# Save export audit
export_df.to_csv(
    "/content/PAPSS2_FINAL_TOP10_SDF_EXPORT_AUDIT.csv",
    index=False
)
```
