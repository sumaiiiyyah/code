```python
# ============================================================
# STEP 4: SDF STRUCTURE VALIDATION USING RDKIT
# ============================================================

from rdkit import Chem
from rdkit.Chem import Descriptors, rdMolDescriptors
import pandas as pd
import os

sdf_file = "/content/PAPSS2_TOP10_COMPOUNDS.sdf"

if not os.path.exists(sdf_file):
    raise FileNotFoundError("Uploaded SDF file not found.")

supplier = Chem.SDMolSupplier(
    sdf_file,
    removeHs=False,
    sanitize=True
)

records = []
invalid_records = []

for index, mol in enumerate(supplier):

    if mol is None:
        invalid_records.append(index + 1)
        continue

    properties = mol.GetPropsAsDict()

    records.append({
        "SDF_Record": index + 1,
        "Molecule_Name": mol.GetProp("_Name") if mol.HasProp("_Name") else "",
        "SDF_Properties": str(properties),
        "Molecular_Formula": rdMolDescriptors.CalcMolFormula(mol),
        "Molecular_Weight": round(Descriptors.MolWt(mol), 4),
        "Canonical_SMILES": Chem.MolToSmiles(
            Chem.RemoveHs(mol),
            canonical=True
        ),
        "InChIKey": Chem.MolToInchiKey(Chem.RemoveHs(mol)),
        "Heavy_Atoms": mol.GetNumHeavyAtoms(),
        "Structure_Valid": True
    })

sdf_audit = pd.DataFrame(records)

print("========== SDF VALIDATION ==========")
print("Total SDF records:", len(records))
print("Valid structures:", len(records))
print("Invalid records:", len(invalid_records))
print("Invalid record numbers:", invalid_records)

print("\nSTRUCTURE DETAILS:")
display(sdf_audit)

# Check structure duplicates
if not sdf_audit.empty:
    print(
        "\nDuplicate structures:",
        sdf_audit["InChIKey"].duplicated().sum()
    )

# Save audit
audit_file = "/content/02_SDF_Structure_Audit.xlsx"

sdf_audit.to_excel(audit_file, index=False)

print("\nSAVED:", audit_file)
```

```python
# ============================================================
# STEP 5: MATCH SDF STRUCTURES WITH ORIGINAL COMPOUND CIDs
# ============================================================

from rdkit import Chem
import pandas as pd

sdf_file = "/content/PAPSS2_TOP10_COMPOUNDS.sdf"
excel_file = "/content/PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"

original = pd.read_excel(excel_file)

supplier = Chem.SDMolSupplier(
    sdf_file,
    removeHs=False,
    sanitize=True
)

matching_records = []

for i, mol in enumerate(supplier):

    if mol is None:
        continue

    props = mol.GetPropsAsDict()

    # Extract original PubChem CID from SDF properties
    cid_value = props.get("PubChem_CID", None)

    matching_records.append({
        "SDF_Record": i + 1,
        "SDF_CID": str(cid_value) if cid_value is not None else None,
        "SDF_Formula": Chem.rdMolDescriptors.CalcMolFormula(mol),
        "SDF_Molecular_Weight": round(
            Chem.Descriptors.MolWt(mol), 4
        ),
        "SDF_SMILES": Chem.MolToSmiles(mol),
        "Isotope_Atoms": sum(
            atom.GetIsotope() > 0 for atom in mol.GetAtoms()
        )
    })

sdf_df = pd.DataFrame(matching_records)

original["PubChem CID"] = original["PubChem CID"].astype(str)
sdf_df["SDF_CID"] = sdf_df["SDF_CID"].astype(str)

comparison = original.merge(
    sdf_df,
    left_on="PubChem CID",
    right_on="SDF_CID",
    how="outer",
    indicator=True
)

comparison["CID_Match"] = comparison["_merge"].eq("both")

print("TOTAL SDF RECORDS:", len(sdf_df))
print("MATCHED CIDs:", comparison["CID_Match"].sum())
print("UNMATCHED RECORDS:", (~comparison["CID_Match"]).sum())

display(comparison[[
    "PubChem CID",
    "PubChem Title",
    "SDF_Record",
    "SDF_CID",
    "Molecular Formula",
    "SDF_Formula",
    "Molecular Weight",
    "SDF_Molecular_Weight",
    "Isotope_Atoms",
    "CID_Match"
]])

comparison.to_excel(
    "/content/03_CID_Structure_Matching.xlsx",
    index=False
)

print("\nSAVED: 03_CID_Structure_Matching.xlsx")
```

```python
# ============================================================
# STEP 7B: ADMET-AI PREDICTIONS FOR ALL 10 COMPOUNDS
# ============================================================

import os
import pandas as pd
import inspect
import admet_ai

from rdkit import Chem
from admet_ai import ADMETModel
from pathlib import Path

# ------------------------------------------------------------
# 1. LOAD ORIGINAL INPUTS
# ------------------------------------------------------------

sdf_file = "/content/PAPSS2_TOP10_COMPOUNDS.sdf"
excel_file = "/content/PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"

if not os.path.exists(sdf_file):
    raise FileNotFoundError("Original SDF file not found.")

if not os.path.exists(excel_file):
    raise FileNotFoundError("Original docking workbook not found.")

original = pd.read_excel(excel_file)

supplier = Chem.SDMolSupplier(
    sdf_file,
    removeHs=False,
    sanitize=True
)

input_records = []

for i, mol in enumerate(supplier):

    if mol is None:
        raise ValueError(f"Invalid SDF structure at record {i+1}")

    props = mol.GetPropsAsDict()

    if "PubChem_CID" not in props:
        raise ValueError(f"CID missing from SDF record {i+1}")

    cid = int(props["PubChem_CID"])

    smiles = Chem.MolToSmiles(
        Chem.RemoveHs(mol),
        canonical=True,
        isomericSmiles=True
    )

    input_records.append({
        "CID": cid,
        "SMILES": smiles,
        "SDF_Record": i + 1
    })

admet_input = pd.DataFrame(input_records)

if len(admet_input) != 10:
    raise ValueError("Expected 10 valid structures.")

if admet_input["CID"].nunique() != 10:
    raise ValueError("Duplicate or missing compound identities.")

# Match compound names from the original workbook
master = original[[
    "PubChem CID",
    "PubChem Title",
    "Edock",
    "Final Rank"
]].copy()

master = master.rename(columns={
    "PubChem CID": "CID",
    "PubChem Title": "Compound"
})

admet_input = admet_input.merge(
    master,
    on="CID",
    how="left",
    validate="one_to_one"
)

if admet_input["Compound"].isna().any():
    raise ValueError("Some SDF CIDs do not match the original workbook.")

# Flag the unresolved isotope discrepancy
admet_input["Structure_Identity_Flag"] = "Structure matched"

admet_input.loc[
    admet_input["CID"] == 49849336,
    "Structure_Identity_Flag"
] = "Isotope labels missing in uploaded SDF; verify"

print("INPUT STRUCTURES VERIFIED:", len(admet_input))
print("UNRESOLVED STRUCTURE FLAGS:",
      (admet_input["Structure_Identity_Flag"] != "Structure matched").sum())

# ------------------------------------------------------------
# 2. RUN ADMET-AI
# ------------------------------------------------------------

model = ADMETModel()

smiles_list = admet_input["SMILES"].tolist()

predictions = model.predict(smiles=smiles_list)

if not isinstance(predictions, pd.DataFrame):
    predictions = pd.DataFrame(predictions)

predictions_fixed = predictions.copy()
predictions_fixed["SMILES"] = predictions_fixed.index.astype(str)
predictions_fixed = predictions_fixed.reset_index(drop=True)

# Prevent accidental duplicate prediction columns
if predictions_fixed["SMILES"].duplicated().any():
    raise ValueError("Duplicate SMILES in ADMET predictions.")

admet_results = admet_input.merge(
    predictions_fixed,
    on="SMILES",
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 3. VALIDATE PREDICTION COMPLETENESS
# ------------------------------------------------------------

prediction_columns = [
    col for col in predictions.columns
    if col not in ["SMILES"]
]

missing_predictions = admet_results[prediction_columns].isna().sum()

print("\n========== PREDICTION AUDIT ==========")
print("Compounds submitted:", len(admet_input))
print("Prediction rows:", len(predictions))
print("Prediction endpoints:", len(prediction_columns))
print("Missing prediction values:", int(missing_predictions.sum()))

if len(predictions) != 10:
    raise ValueError("ADMET-AI did not return 10 prediction rows.")

if missing_predictions.sum() > 0:
    raise ValueError("Missing ADMET predictions detected.")

# ------------------------------------------------------------
# 4. LOAD ENDPOINT METADATA
# ------------------------------------------------------------

package_path = Path(inspect.getfile(admet_ai)).parent
metadata_file = package_path / "resources" / "data" / "admet.csv"

if not metadata_file.exists():
    raise FileNotFoundError("ADMET endpoint metadata file not found.")

endpoint_metadata = pd.read_csv(metadata_file)

# ------------------------------------------------------------
# 5. SEPARATE TOXICITY AND OTHER ADMET ENDPOINTS
# ------------------------------------------------------------

toxicity_endpoints = [
    "AMES",
    "Carcinogens_Lagunin",
    "ClinTox",
    "DILI",
    "hERG",
    "Skin_Reaction",
    "NR-AR",
    "NR-AR-LBD",
    "NR-AhR",
    "NR-Aromatase",
    "NR-ER",
    "NR-ER-LBD",
    "NR-PPAR-gamma",
    "SR-ARE",
    "SR-ATAD5",
    "SR-HSE",
    "SR-MMP",
    "SR-p53",
    "LD50_Zhu"
]

available_toxicity = [
    col for col in toxicity_endpoints
    if col in admet_results.columns
]

toxicity_results = admet_results[
    ["CID", "Compound", "Structure_Identity_Flag"]
    + available_toxicity
].copy()

other_admet_columns = [
    col for col in prediction_columns
    if col not in available_toxicity
]

other_admet_results = admet_results[
    ["CID", "Compound", "Structure_Identity_Flag"]
    + other_admet_columns
].copy()

# ------------------------------------------------------------
# 6. SAVE COMPLETE RESULTS
# ------------------------------------------------------------

output_file = "/content/05_PAPSS2_ADMET_Toxicity_All10.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    admet_results.to_excel(
        writer,
        sheet_name="All_Raw_Predictions",
        index=False
    )

    toxicity_results.to_excel(
        writer,
        sheet_name="Toxicity_Predictions",
        index=False
    )

    other_admet_results.to_excel(
        writer,
        sheet_name="Other_ADMET",
        index=False
    )

    endpoint_metadata.to_excel(
        writer,
        sheet_name="Endpoint_Metadata",
        index=False
    )

    admet_input.to_excel(
        writer,
        sheet_name="Structure_Input_Audit",
        index=False
    )

# ------------------------------------------------------------
# 7. DISPLAY VALIDATION
# ------------------------------------------------------------

print("\n========== FINAL RESULTS ==========")
print("ADMET-AI version:", admet_ai.__version__)
print("Total compounds:", len(admet_results))
print("Toxicity endpoints available:", len(available_toxicity))
print("Other ADMET endpoints:", len(other_admet_columns))
print("Missing prediction values:", int(missing_predictions.sum()))

print("\nTOXICITY PREDICTIONS:")
display(toxicity_results)

print("\nOTHER ADMET PREDICTIONS:")
display(other_admet_results)

print("\nSAVED WORKBOOK:", output_file)
```

```python
import os
import time
import requests
import pandas as pd
from rdkit import Chem
from openpyxl import load_workbook

# ------------------------------------------------------------
# 1. LOAD THE AUTHORITATIVE COMPOUND LIST AND SDF
# ------------------------------------------------------------

excel_path = "/content/PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"
sdf_path = "/content/PAPSS2_TOP10_COMPOUNDS.sdf"

assert os.path.exists(excel_path), "Docking Excel file not found."
assert os.path.exists(sdf_path), "SDF file not found."

df = pd.read_excel(excel_path)

supplier = Chem.SDMolSupplier(sdf_path, removeHs=False)
sdf_records = {}

for mol in supplier:
    if mol is None:
        continue

    props = mol.GetPropsAsDict()
    cid = str(props.get("PubChem_CID", "")).strip()

    if cid:
        sdf_records[cid] = props

# Identify the relevant columns without changing the source data.
cid_col = "PubChem CID"
name_col = "PubChem Title"

assert cid_col in df.columns, f"Missing column: {cid_col}"
assert name_col in df.columns, f"Missing column: {name_col}"

# ------------------------------------------------------------
# 2. BUILD SEARCH TASKS
# ------------------------------------------------------------

tasks = []

for _, row in df.iterrows():

    cid = str(row[cid_col]).strip()
    compound_name = str(row[name_col]).strip()

    props = sdf_records.get(cid, {})
    parent_ligand = str(props.get("Parent_Ligand", "")).strip()

    if not compound_name or compound_name.lower() == "nan":
        compound_name = str(props.get("PubChem_Title", "")).strip()

    # Exact compound literature
    if compound_name and compound_name.lower() != "nan":
        tasks.append({
            "CID": cid,
            "Compound": compound_name,
            "Evidence_Type": "Exact compound",
            "Query": f'"{compound_name}"'
        })

    # Scaffold / parent-ligand literature
    if parent_ligand and parent_ligand.lower() != "nan":
        tasks.append({
            "CID": cid,
            "Compound": compound_name,
            "Evidence_Type": "Scaffold anticancer",
            "Query": f'"{parent_ligand}" anticancer OR cytotoxicity OR cancer'
        })

        tasks.append({
            "CID": cid,
            "Compound": compound_name,
            "Evidence_Type": "Scaffold breast cancer",
            "Query": f'"{parent_ligand}" breast cancer OR MCF-7 OR MDA-MB-231'
        })

# PAPSS2 target-specific evidence
tasks.append({
    "CID": "",
    "Compound": "PAPSS2",
    "Evidence_Type": "PAPSS2 breast cancer",
    "Query": "PAPSS2 breast cancer"
})

tasks.append({
    "CID": "",
    "Compound": "PAPSS2",
    "Evidence_Type": "PAPSS2 cancer mechanism",
    "Query": "PAPSS2 cancer sulfation tumour"
})

print("COMPOUNDS:", len(df))
print("SDF RECORDS:", len(sdf_records))
print("LITERATURE SEARCH TASKS:", len(tasks))
print("SCAFFOLD FIELDS FOUND:", sum(
    bool(str(x.get("Parent_Ligand", "")).strip())
    for x in sdf_records.values()
))

# ------------------------------------------------------------
# 3. OPENALEX SEARCH
# ------------------------------------------------------------

session = requests.Session()
session.headers.update({
    "User-Agent": "PAPSS2-research-literature-audit/1.0"
})

records = []
search_log = []

def reconstruct_abstract(inverted_index):
    if not inverted_index:
        return ""

    words = []
    for word, positions in inverted_index.items():
        for position in positions:
            words.append((position, word))

    words.sort()
    return " ".join(word for _, word in words)

for i, task in enumerate(tasks, start=1):

    query = task["Query"]
    success = False
    error_message = ""
    results_count = 0

    for attempt in range(3):
        try:
            response = session.get(
                "https://api.openalex.org/works",
                params={
                    "search": query,
                    "per-page": 10
                },
                timeout=40
            )

            response.raise_for_status()
            results = response.json().get("results", [])
            results_count = len(results)
            success = True
            break

        except Exception as error:
            error_message = str(error)
            time.sleep(2 * (attempt + 1))

    search_log.append({
        "Search_Number": i,
        "CID": task["CID"],
        "Compound": task["Compound"],
        "Evidence_Type": task["Evidence_Type"],
        "Query": query,
        "Successful": success,
        "Results_Returned": results_count,
        "Error": error_message
    })

    if success:
        for item in results:
            records.append({
                "CID": task["CID"],
                "Compound": task["Compound"],
                "Evidence_Type": task["Evidence_Type"],
                "Search_Query": query,
                "OpenAlex_ID": item.get("id", ""),
                "Title": item.get("title", ""),
                "Year": item.get("publication_year", ""),
                "DOI": item.get("doi", ""),
                "Type": item.get("type", ""),
                "Journal": (
                    item.get("primary_location") or {}
                ).get("source", {}).get("display_name", ""),
                "Authors": "; ".join([
                    a.get("author", {}).get("display_name", "")
                    for a in item.get("authorships", [])[:8]
                ]),
                "Abstract": reconstruct_abstract(
                    item.get("abstract_inverted_index")
                ),
                "Landing_Page": item.get("id", "")
            })

    print(
        f"{i}/{len(tasks)} | {task['Evidence_Type']} | "
        f"{task['Compound'][:55]} | Results: {results_count}"
    )

    time.sleep(0.25)

# ------------------------------------------------------------
# 4. ORGANISE AND DEDUPLICATE RESULTS
# ------------------------------------------------------------

literature = pd.DataFrame(records)
log_df = pd.DataFrame(search_log)

if not literature.empty:
    literature = literature.drop_duplicates(
        subset=["OpenAlex_ID", "Evidence_Type", "CID"]
    ).reset_index(drop=True)

    literature = literature.sort_values(
        ["Evidence_Type", "CID", "Year"],
        ascending=[True, True, False],
        na_position="last"
    )

# Candidate-level search summary
if not literature.empty:
    summary = (
        literature.groupby(["CID", "Compound", "Evidence_Type"])
        .agg(
            Records_Retrieved=("OpenAlex_ID", "nunique"),
            Publication_Years=(
                "Year",
                lambda x: "; ".join(
                    sorted(set(str(v) for v in x if pd.notna(v) and str(v))
                    , reverse=True)
                )
            )
        )
        .reset_index()
    )
else:
    summary = pd.DataFrame(columns=[
        "CID", "Compound", "Evidence_Type",
        "Records_Retrieved", "Publication_Years"
    ])

# ------------------------------------------------------------
# 5. SAVE THE LITERATURE AUDIT
# ------------------------------------------------------------

output_path = "/content/06_PAPSS2_Literature_Search.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    literature.to_excel(
        writer, sheet_name="Literature_Records", index=False
    )
    log_df.to_excel(
        writer, sheet_name="Search_Log", index=False
    )
    summary.to_excel(
        writer, sheet_name="Candidate_Summary", index=False
    )

print("\nLITERATURE SEARCH COMPLETED")
print("Successful searches:", int(log_df["Successful"].sum()))
print("Failed searches:", int((~log_df["Successful"]).sum()))
print("Unique literature records:", len(literature))
print("Output saved:", output_path)
```

```python
import requests
import pandas as pd
import time
import os

# Load previous search log
old_log = pd.read_excel(
    "/content/06_PAPSS2_Literature_Search.xlsx",
    sheet_name="Search_Log"
)

# Retry only searches blocked by HTTP errors
retry = old_log[
    old_log["Error"].fillna("").str.contains("429|500|502|503|504")
].copy()

print("SEARCHES TO RECOVER:", len(retry))

session = requests.Session()
session.headers.update({
    "User-Agent": "PAPSS2-literature-review/1.0 (academic research)"
})

crossref_records = []
crossref_log = []

for i, (_, task) in enumerate(retry.iterrows(), start=1):

    query = str(task["Query"])
    success = False
    error = ""
    items = []

    for attempt in range(3):
        try:
            response = session.get(
                "https://api.crossref.org/works",
                params={
                    "query.bibliographic": query,
                    "rows": 10
                },
                timeout=40
            )

            response.raise_for_status()
            items = response.json()["message"]["items"]
            success = True
            break

        except Exception as e:
            error = str(e)
            time.sleep(3 * (attempt + 1))

    crossref_log.append({
        "Original_Search_Number": task["Search_Number"],
        "CID": task["CID"],
        "Compound": task["Compound"],
        "Evidence_Type": task["Evidence_Type"],
        "Query": query,
        "Successful": success,
        "Results_Returned": len(items),
        "Error": error
    })

    for item in items:
        authors = item.get("author", [])

        crossref_records.append({
            "CID": task["CID"],
            "Compound": task["Compound"],
            "Evidence_Type": task["Evidence_Type"],
            "Search_Query": query,
            "Title": "; ".join(item.get("title", [])),
            "Year": (
                item.get("published-print", {}).get("date-parts", [[]])[0][0]
                if item.get("published-print", {}).get("date-parts")
                else (
                    item.get("published-online", {}).get("date-parts", [[]])[0][0]
                    if item.get("published-online", {}).get("date-parts")
                    else ""
                )
            ),
            "DOI": item.get("DOI", ""),
            "Journal": (
                item.get("container-title", [""])[0]
                if item.get("container-title") else ""
            ),
            "Authors": "; ".join(
                a.get("family", "") for a in authors[:8]
            ),
            "URL": item.get("URL", "")
        })

    crossref_log[-1]["Results_Returned"] = len(items)

    print(
        f"{i}/{len(retry)} | {task['Evidence_Type']} | "
        f"Results: {len(items)} | Success: {success}"
    )

    time.sleep(1.5)

# Save recovery results separately
records_df = pd.DataFrame(crossref_records)
log_df = pd.DataFrame(crossref_log)

output_path = "/content/07_PAPSS2_Crossref_Literature_Recovery.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    records_df.to_excel(
        writer, sheet_name="Crossref_Records", index=False
    )
    log_df.to_excel(
        writer, sheet_name="Recovery_Search_Log", index=False
    )

print("\nCROSSREF RECOVERY COMPLETED")
print("Searches attempted:", len(retry))
print("Successful requests:", int(log_df["Successful"].sum()))
print("Records retrieved:", len(records_df))
print("Saved:", output_path)
```

```python
import pandas as pd
import os

openalex_path = "/content/06_PAPSS2_Literature_Search.xlsx"
crossref_path = "/content/07_PAPSS2_Crossref_Literature_Recovery.xlsx"

oa = pd.read_excel(openalex_path, sheet_name="Literature_Records")
cr = pd.read_excel(crossref_path, sheet_name="Crossref_Records")

# Standardise columns
oa = oa.rename(columns={
    "OpenAlex_ID": "Database_ID",
    "Landing_Page": "URL"
})
oa["Source_Database"] = "OpenAlex"

cr = cr.rename(columns={
    "URL": "URL"
})
cr["Database_ID"] = ""
cr["Source_Database"] = "Crossref"

# Combine common fields
common_columns = [
    "CID", "Compound", "Evidence_Type", "Search_Query",
    "Title", "Year", "DOI", "Journal", "Authors",
    "Abstract", "URL", "Database_ID", "Source_Database"
]

for col in common_columns:
    if col not in oa.columns:
        oa[col] = ""
    if col not in cr.columns:
        cr[col] = ""

master = pd.concat(
    [oa[common_columns], cr[common_columns]],
    ignore_index=True
)

# Normalise DOI for deduplication
master["DOI_Normalised"] = (
    master["DOI"]
    .fillna("")
    .astype(str)
    .str.lower()
    .str.replace("https://doi.org/", "", regex=False)
    .str.replace("http://doi.org/", "", regex=False)
    .str.strip()
)

# Deduplicate by DOI; retain records without DOI separately
with_doi = master[master["DOI_Normalised"] != ""].drop_duplicates(
    subset=["DOI_Normalised", "Evidence_Type", "CID"]
)

without_doi = master[master["DOI_Normalised"] == ""]

master = pd.concat(
    [with_doi, without_doi],
    ignore_index=True
)

# Keep a DOI-level bibliography for review
bibliography = master.drop_duplicates(
    subset=["DOI_Normalised"],
    keep="first"
)

# Summary by candidate and evidence category
candidate_summary = (
    master.groupby(["CID", "Compound", "Evidence_Type"])
    .agg(
        Retrieved_Records=("Title", "count"),
        Unique_DOIs=("DOI_Normalised", lambda x: x[x != ""].nunique())
    )
    .reset_index()
)

output_path = "/content/08_PAPSS2_Master_Literature.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    master.to_excel(writer, sheet_name="All_Search_Records", index=False)
    bibliography.to_excel(writer, sheet_name="DOI_Bibliography", index=False)
    candidate_summary.to_excel(writer, sheet_name="Candidate_Summary", index=False)

print("MASTER LITERATURE AUDIT")
print("=======================")
print("OpenAlex records:", len(oa))
print("Crossref records:", len(cr))
print("Combined records before deduplication:", len(oa) + len(cr))
print("Records after deduplication:", len(master))
print("Unique DOI bibliography:", len(bibliography))
print("Candidates represented:", master["CID"].nunique())
print("Saved:", output_path)
```

```python
import pandas as pd
import re

input_path = "/content/08_PAPSS2_Master_Literature.xlsx"

df = pd.read_excel(input_path, sheet_name="All_Search_Records")

# Normalise DOI and title
df["DOI_Normalised"] = (
    df["DOI"].fillna("").astype(str).str.lower()
    .str.replace("https://doi.org/", "", regex=False)
    .str.replace("http://doi.org/", "", regex=False)
    .str.strip()
)

def normalise_title(value):
    value = str(value).lower()
    value = re.sub(r"[^a-z0-9]+", " ", value)
    return re.sub(r"\s+", " ", value).strip()

df["Title_Normalised"] = df["Title"].apply(normalise_title)

# Deduplicate records using DOI where available;
# otherwise use the normalised title.
df["Record_Key"] = df.apply(
    lambda r: (
        "doi:" + r["DOI_Normalised"]
        if r["DOI_Normalised"]
        else "title:" + r["Title_Normalised"]
    ),
    axis=1
)

# One master record per publication.
# Preserve all associated evidence categories, CIDs and databases.
master = (
    df.groupby("Record_Key", dropna=False)
    .agg(
        Title=("Title", "first"),
        Year=("Year", "first"),
        DOI=("DOI", "first"),
        Journal=("Journal", "first"),
        Authors=("Authors", "first"),
        Abstract=("Abstract", "first"),
        URL=("URL", "first"),
        Evidence_Categories=(
            "Evidence_Type",
            lambda x: "; ".join(sorted(set(str(v) for v in x if pd.notna(v))))
        ),
        Associated_CIDs=(
            "CID",
            lambda x: "; ".join(sorted(set(str(v) for v in x if pd.notna(v) and str(v) != "")))
        ),
        Databases=(
            "Source_Database",
            lambda x: "; ".join(sorted(set(str(v) for v in x if pd.notna(v))))
        ),
        Search_Queries=(
            "Search_Query",
            lambda x: " | ".join(sorted(set(str(v) for v in x if pd.notna(v))))
        )
    )
    .reset_index()
)

# Preliminary keyword flags based on title and abstract only.
# These are screening aids, NOT confirmed relevance classifications.
def flag_relevance(row):
    text = (
        str(row.get("Title", "")) + " " +
        str(row.get("Abstract", ""))
    ).lower()

    breast = any(x in text for x in [
        "breast cancer", "breast carcinoma", "mcf-7",
        "mda-mb-231", "mda mb 231", "triple-negative breast"
    ])

    cancer = any(x in text for x in [
        "anticancer", "anti-cancer", "cancer", "carcinoma",
        "tumor", "tumour", "cytotoxic", "neoplasm"
    ])

    papss2 = "papss2" in text or "3'-phosphoadenosine 5'-phosphosulfate synthetase 2" in text

    return pd.Series({
        "Title_Abstract_Breast_Cancer_Term": breast,
        "Title_Abstract_Cancer_Term": cancer,
        "Title_Abstract_PAPSS2_Term": papss2
    })

flags = master.apply(flag_relevance, axis=1)
master = pd.concat([master, flags], axis=1)

# Add fields for manual verification
master["Exact_Compound_Evidence_Confirmed"] = ""
master["Scaffold_Evidence_Confirmed"] = ""
master["Breast_Cancer_Evidence_Confirmed"] = ""
master["Experimental_Model"] = ""
master["Main_Finding"] = ""
master["Include_in_Review"] = ""
master["Reviewer_Notes"] = ""

# Summary
summary = pd.DataFrame({
    "Metric": [
        "Original combined records",
        "Unique publication records",
        "Records with DOI",
        "Records without DOI",
        "Records with breast-cancer terms",
        "Records with cancer-related terms",
        "Records with PAPSS2 terms"
    ],
    "Count": [
        len(df),
        len(master),
        master["DOI"].fillna("").astype(str).str.strip().ne("").sum(),
        master["DOI"].fillna("").astype(str).str.strip().eq("").sum(),
        master["Title_Abstract_Breast_Cancer_Term"].sum(),
        master["Title_Abstract_Cancer_Term"].sum(),
        master["Title_Abstract_PAPSS2_Term"].sum()
    ]
})

output_path = "/content/09_PAPSS2_Literature_Screening.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    master.to_excel(writer, sheet_name="Unique_Publications", index=False)
    summary.to_excel(writer, sheet_name="Screening_Summary", index=False)

print("LITERATURE SCREENING PREPARATION")
print("================================")
print(summary.to_string(index=False))
print("\nSaved:", output_path)
```

```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, QED, rdMolDescriptors
from rdkit.Chem.FilterCatalog import FilterCatalog, FilterCatalogParams

sdf_path = "/content/PAPSS2_TOP10_COMPOUNDS.sdf"
excel_path = "/content/PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"

supplier = Chem.SDMolSupplier(sdf_path, removeHs=False)
docking = pd.read_excel(excel_path)

# PAINS filter configuration
params = FilterCatalogParams()
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_A)
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_B)
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_C)

pains_catalog = FilterCatalog(params)

results = []

for mol in supplier:

    if mol is None:
        continue

    props = mol.GetPropsAsDict()
    cid = str(props.get("PubChem_CID", "")).strip()

    # Use a hydrogen-stripped copy for descriptor calculations
    mol_clean = Chem.RemoveHs(mol)

    if not mol_clean.GetNumAtoms():
        continue

    # Molecular descriptors
    mw = Descriptors.MolWt(mol_clean)
    logp = Descriptors.MolLogP(mol_clean)
    tpsa = rdMolDescriptors.CalcTPSA(mol_clean)

    hbd = Lipinski.NumHDonors(mol_clean)
    hba = Lipinski.NumHAcceptors(mol_clean)
    rotatable = Lipinski.NumRotatableBonds(mol_clean)

    rings = rdMolDescriptors.CalcNumRings(mol_clean)
    aromatic_rings = rdMolDescriptors.CalcNumAromaticRings(mol_clean)
    fraction_csp3 = rdMolDescriptors.CalcFractionCSP3(mol_clean)

    heavy_atoms = mol_clean.GetNumHeavyAtoms()
    formal_charge = Chem.GetFormalCharge(mol_clean)

    qed_value = QED.qed(mol_clean)

    # Lipinski violations
    violations = sum([
        mw > 500,
        logp > 5,
        hbd > 5,
        hba > 10
    ])

    # PAINS structural alerts
    pains_matches = pains_catalog.GetMatches(mol_clean)
    pains_alerts = "; ".join(
        sorted(set(match.GetDescription() for match in pains_matches))
    )

    results.append({
        "PubChem_CID": cid,
        "Molecular_Weight_RDKit": mw,
        "cLogP_RDKit": logp,
        "TPSA": tpsa,
        "HBD": hbd,
        "HBA": hba,
        "Rotatable_Bonds": rotatable,
        "Ring_Count": rings,
        "Aromatic_Ring_Count": aromatic_rings,
        "Fraction_CSP3": fraction_csp3,
        "Heavy_Atom_Count": heavy_atoms,
        "Formal_Charge": formal_charge,
        "QED": qed_value,
        "Lipinski_Violations": violations,
        "PAINS_Alert": pains_alerts if pains_alerts else "None detected",
        "SDF_Isotope_Atoms": sum(
            atom.GetIsotope() != 0 for atom in mol_clean.GetAtoms()
        ),
        "Structure_Identity_Flag": (
            "VERIFY: isotope-labelled workbook candidate; SDF is non-labelled"
            if cid == "49849336"
            else "SDF structure used"
        )
    })

descriptors = pd.DataFrame(results)

# Preserve docking identifiers and scores
docking["PubChem CID"] = docking["PubChem CID"].astype(str).str.strip()

merged = docking.merge(
    descriptors,
    left_on="PubChem CID",
    right_on="PubChem_CID",
    how="left",
    validate="one_to_one"
)

output_path = "/content/13_PAPSS2_Molecular_Properties.xlsx"

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    merged.to_excel(writer, sheet_name="Molecular_Properties", index=False)
    descriptors.to_excel(writer, sheet_name="RDKit_Descriptors", index=False)

print("MOLECULAR PROPERTY ASSESSMENT")
print("============================")
print("SDF structures processed:", len(descriptors))
print("Matched docking compounds:", merged["PubChem_CID"].notna().sum())
print("Missing descriptor records:", merged["PubChem_CID"].isna().sum())
print("PAINS alerts detected:",
      (descriptors["PAINS_Alert"] != "None detected").sum())
print("Saved:", output_path)

print("\nDESCRIPTOR PREVIEW:")
print(
    merged[[
        "PubChem CID",
        "Molecular_Weight_RDKit",
        "cLogP_RDKit",
        "TPSA",
        "QED",
        "Lipinski_Violations",
        "PAINS_Alert"
    ]].to_string(index=False)
)
```

```python
import sys
import subprocess
import importlib.util
from pathlib import Path

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
INPUT = BASE / "01_Input_Files"
OUTPUT = BASE / "02_Analysis_Results"
OUTPUT.mkdir(parents=True, exist_ok=True)

# Install dependencies if missing
if importlib.util.find_spec("admet_ai") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "admet-ai"
    ])

if importlib.util.find_spec("rdkit") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "rdkit"
    ])

import pandas as pd
from rdkit import Chem
from admet_ai import ADMETModel

sdf_path = INPUT / "PAPSS2_TOP10_COMPOUNDS.sdf"
excel_path = INPUT / "PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"

if not sdf_path.exists() or not excel_path.exists():
    raise FileNotFoundError("One or both source files are missing.")

# Read authoritative docking workbook
docking = pd.read_excel(excel_path)
docking["PubChem CID"] = (
    docking["PubChem CID"].astype(str).str.replace(r"\.0$", "", regex=True).str.strip()
)

# Extract valid structures and identifiers
supplier = Chem.SDMolSupplier(str(sdf_path), removeHs=False)

records = []
for index, mol in enumerate(supplier):
    if mol is None:
        print(f"WARNING: Invalid SDF record at position {index + 1}")
        continue

    props = mol.GetPropsAsDict()
    cid = str(props.get("PubChem_CID", "")).replace(".0", "").strip()

    if not cid:
        print(f"WARNING: Missing PubChem_CID at SDF record {index + 1}")
        continue

    clean_mol = Chem.RemoveHs(mol)
    smiles = Chem.MolToSmiles(clean_mol, isomericSmiles=True)

    records.append({
        "PubChem CID": cid,
        "SMILES": smiles,
        "SDF_Record": index + 1,
        "Isotope_Atoms": sum(
            atom.GetIsotope() != 0 for atom in clean_mol.GetAtoms()
        )
    })

structures = pd.DataFrame(records)

if structures["PubChem CID"].duplicated().any():
    raise ValueError("Duplicate CIDs found in SDF. Resolve before prediction.")

merged = docking.merge(
    structures,
    on="PubChem CID",
    how="left",
    validate="one_to_one"
)

if merged["SMILES"].isna().any():
    raise ValueError("Some docking compounds could not be matched to SDF structures.")

print("Matched structures:", len(merged))
print("Running ADMET-AI predictions...")

model = ADMETModel()
predictions = model.predict(smiles=merged["SMILES"].tolist())

predictions = predictions.reset_index(drop=True)
predictions.insert(0, "PubChem CID", merged["PubChem CID"].values)

# Combine docking identifiers, scores and predictions
combined = pd.concat([
    merged.reset_index(drop=True),
    predictions.drop(columns=["PubChem CID"], errors="ignore")
], axis=1)

# Preserve the known isotope identity warning
combined["Structure_Identity_Flag"] = combined["PubChem CID"].apply(
    lambda x: (
        "WARNING: isotope-labelled workbook identity does not match unlabelled SDF"
        if str(x) == "49849336" else "No discrepancy identified in previous audit"
    )
)

toxicity_cols = [
    c for c in predictions.columns
    if c != "PubChem CID" and (
        "tox" in c.lower()
        or "ames" in c.lower()
        or "herg" in c.lower()
        or "skin" in c.lower()
        or "ld50" in c.lower()
        or c.lower().startswith("sr-")
    )
]

other_cols = [
    c for c in predictions.columns
    if c not in ["PubChem CID"] + toxicity_cols
]

audit = pd.DataFrame({
    "Metric": [
        "SDF records processed",
        "Matched docking compounds",
        "Prediction endpoints returned",
        "Missing prediction values",
        "Toxicity-related columns identified"
    ],
    "Value": [
        len(structures),
        len(merged),
        len(predictions.columns) - 1,
        int(predictions.drop(columns=["PubChem CID"], errors="ignore").isna().sum().sum()),
        len(toxicity_cols)
    ]
})

output_file = OUTPUT / "05_PAPSS2_ADMET_Toxicity_All10.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    combined.to_excel(writer, sheet_name="All_Predictions", index=False)
    combined[["PubChem CID"] + toxicity_cols].to_excel(
        writer, sheet_name="Toxicity_Predictions", index=False
    )
    combined[["PubChem CID"] + other_cols].to_excel(
        writer, sheet_name="Other_ADMET", index=False
    )
    audit.to_excel(writer, sheet_name="Prediction_Audit", index=False)

print("\nADMET-AI RESTORATION COMPLETE")
print("=============================")
print(audit.to_string(index=False))
print("\nSaved:", output_file)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, QED, rdMolDescriptors
from rdkit.Chem.FilterCatalog import FilterCatalog, FilterCatalogParams

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
INPUT = BASE / "01_Input_Files"
OUT = BASE / "02_Analysis_Results"
OUT.mkdir(parents=True, exist_ok=True)

excel = INPUT / "PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"
sdf = INPUT / "PAPSS2_TOP10_COMPOUNDS.sdf"

assert excel.exists() and sdf.exists(), "Source file missing."

# 1. Reconstruct authoritative docking results
docking = pd.read_excel(excel)
docking["PubChem CID"] = (
    docking["PubChem CID"].astype(str)
    .str.replace(r"\.0$", "", regex=True).str.strip()
)
docking["Edock"] = pd.to_numeric(docking["Edock"], errors="coerce")

if docking["PubChem CID"].duplicated().any():
    raise ValueError("Duplicate CIDs in docking workbook.")
if docking["Edock"].isna().any():
    raise ValueError("Missing or invalid Edock scores.")

docking["Edock_Rank_Recalculated"] = (
    docking["Edock"].rank(method="min", ascending=True).astype(int)
)
docking = docking.sort_values(
    ["Edock", "PubChem CID"], ascending=[True, True]
).reset_index(drop=True)

docking.to_excel(
    OUT / "04_Authoritative_Docking_Results.xlsx", index=False
)

# 2. Recalculate RDKit descriptors
params = FilterCatalogParams()
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_A)
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_B)
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS_C)
pains = FilterCatalog(params)

records = []
for mol in Chem.SDMolSupplier(str(sdf), removeHs=False):
    if mol is None:
        continue

    props = mol.GetPropsAsDict()
    cid = str(props.get("PubChem_CID", "")).replace(".0", "").strip()
    m = Chem.RemoveHs(mol)

    if not cid or m.GetNumAtoms() == 0:
        continue

    mw = Descriptors.MolWt(m)
    logp = Descriptors.MolLogP(m)
    hbd = Lipinski.NumHDonors(m)
    hba = Lipinski.NumHAcceptors(m)
    alerts = pains.GetMatches(m)

    records.append({
        "PubChem_CID": cid,
        "Molecular_Weight_RDKit": mw,
        "cLogP_RDKit": logp,
        "TPSA": rdMolDescriptors.CalcTPSA(m),
        "HBD": hbd,
        "HBA": hba,
        "Rotatable_Bonds": Lipinski.NumRotatableBonds(m),
        "Rings": rdMolDescriptors.CalcNumRings(m),
        "Aromatic_Rings": rdMolDescriptors.CalcNumAromaticRings(m),
        "Fraction_CSP3": rdMolDescriptors.CalcFractionCSP3(m),
        "Heavy_Atoms": m.GetNumHeavyAtoms(),
        "Formal_Charge": Chem.GetFormalCharge(m),
        "QED": QED.qed(m),
        "Lipinski_Violations": sum([mw > 500, logp > 5, hbd > 5, hba > 10]),
        "PAINS_Alert": "; ".join(sorted(set(
            x.GetDescription() for x in alerts
        ))) if alerts else "None detected"
    })

desc = pd.DataFrame(records)
desc["PubChem_CID"] = desc["PubChem_CID"].astype(str)

if desc["PubChem_CID"].duplicated().any():
    raise ValueError("Duplicate CIDs in SDF descriptor records.")

docking["PubChem CID"] = docking["PubChem CID"].astype(str)
merged = docking.merge(
    desc, left_on="PubChem CID", right_on="PubChem_CID",
    how="left", validate="one_to_one"
)

if merged["Molecular_Weight_RDKit"].isna().any():
    raise ValueError("Some compounds did not match descriptor records.")

merged["Structure_Identity_Flag"] = np.where(
    merged["PubChem CID"] == "49849336",
    "SDF lacks isotope labels; descriptors represent non-labelled structure",
    "No isotope discrepancy identified in previous audit"
)

merged.to_excel(
    OUT / "13_PAPSS2_Molecular_Properties.xlsx", index=False
)

# 3. Reconstruct property flags
flags = merged.copy()
flags["High_Molecular_Weight"] = flags["Molecular_Weight_RDKit"] > 500
flags["High_Lipophilicity"] = flags["cLogP_RDKit"] > 5
flags["Very_Low_TPSA"] = flags["TPSA"] < 20
flags["Low_QED"] = flags["QED"] < 0.30
flags["Lipinski_Violation_Flag"] = flags["Lipinski_Violations"] > 0
flags["PAINS_Flag"] = flags["PAINS_Alert"] != "None detected"

flag_cols = [
    "High_Molecular_Weight", "High_Lipophilicity",
    "Very_Low_TPSA", "Low_QED",
    "Lipinski_Violation_Flag", "PAINS_Flag"
]
flags["Total_Property_Flags"] = flags[flag_cols].sum(axis=1)

summary = flags[[
    "PubChem CID", "Molecular_Weight_RDKit", "cLogP_RDKit",
    "TPSA", "QED", "Lipinski_Violations",
    "Total_Property_Flags", "Structure_Identity_Flag"
]]

with pd.ExcelWriter(
    OUT / "14_PAPSS2_Property_Flag_Assessment.xlsx",
    engine="openpyxl"
) as writer:
    flags.to_excel(writer, sheet_name="Detailed_Property_Flags", index=False)
    summary.to_excel(writer, sheet_name="Summary", index=False)

print("RECONSTRUCTION COMPLETE")
print("=======================")
print("Docking compounds:", len(docking))
print("Matched structures:", len(merged))
print("Missing descriptor records:", merged["QED"].isna().sum())
print("\nReconstructed files:")
for name in [
    "04_Authoritative_Docking_Results.xlsx",
    "13_PAPSS2_Molecular_Properties.xlsx",
    "14_PAPSS2_Property_Flag_Assessment.xlsx"
]:
    print(name, "|", (OUT / name).stat().st_size, "bytes")
```

```python
import pandas as pd
import requests
import time
from pathlib import Path

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
R = BASE / "02_Analysis_Results"

docking = pd.read_excel(R / "04_Authoritative_Docking_Results.xlsx")

def clean_cid(x):
    return str(x).replace(".0", "").strip()

docking["PubChem CID"] = docking["PubChem CID"].apply(clean_cid)

session = requests.Session()
session.headers.update({
    "User-Agent": "PAPSS2-research-literature-audit/1.0 (academic research)"
})

records = []
errors = []

for _, row in docking.iterrows():
    cid = row["PubChem CID"]
    name = str(row.get("PubChem Title", "")).strip()

    queries = [f'"{cid}"']
    if name and name.lower() != "nan":
        queries.append(f'"{name}"')

    for query in queries:
        try:
            response = session.get(
                "https://api.crossref.org/works",
                params={"query.bibliographic": query, "rows": 5},
                timeout=30
            )

            response.raise_for_status()
            items = response.json()["message"].get("items", [])

            for item in items:
                records.append({
                    "PubChem_CID": cid,
                    "Candidate_Name": name,
                    "Search_Query": query,
                    "Paper_Title": "; ".join(item.get("title", [])),
                    "DOI": item.get("DOI", ""),
                    "Year": item.get("published", {}).get(
                        "date-parts", [[None]]
                    )[0][0],
                    "Journal": item.get("container-title", [""])[0],
                    "Abstract": item.get("abstract", ""),
                    "Crossref_Type": item.get("type", "")
                })

            time.sleep(0.5)

        except Exception as e:
            errors.append({
                "PubChem_CID": cid,
                "Search_Query": query,
                "Error": str(e)
            })

columns = [
    "PubChem_CID", "Candidate_Name", "Search_Query",
    "Paper_Title", "DOI", "Year", "Journal",
    "Abstract", "Crossref_Type"
]

papers = pd.DataFrame(records, columns=columns)

if not papers.empty:
    papers = papers.drop_duplicates(
        subset=["PubChem_CID", "DOI", "Paper_Title"]
    )

error_df = pd.DataFrame(
    errors, columns=["PubChem_CID", "Search_Query", "Error"]
)

output = R / "19_PAPSS2_Exact_Compound_Literature_Search.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    papers.to_excel(writer, sheet_name="Search_Results", index=False)
    error_df.to_excel(writer, sheet_name="Search_Errors", index=False)

print("EXACT-COMPOUND LITERATURE SEARCH")
print("================================")
print("Candidates searched:", docking["PubChem CID"].nunique())
print("Literature records retrieved:", len(papers))
print("Unique DOI records:", papers["DOI"].replace("", pd.NA).nunique())
print("Failed requests:", len(error_df))
print("Saved:", output)
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
R = BASE / "02_Analysis_Results"

file = R / "19_PAPSS2_Exact_Compound_Literature_Search.xlsx"
papers = pd.read_excel(file, sheet_name="Search_Results")

def screen(row):
    title = str(row.get("Paper_Title", "")).lower()
    abstract = str(row.get("Abstract", "")).lower()
    text = title + " " + abstract

    cancer_terms = [
        "cancer", "carcinoma", "tumor", "tumour",
        "anticancer", "antitumor", "cytotoxic"
    ]

    compound_terms = [
        "153531369", "153531370", "175018237",
        "150310321", "58965179", "21260039",
        "68870353", "23187967", "14157051", "49849336"
    ]

    scaffold_terms = [
        "thiophene", "benzimidazole", "benzoxazole",
        "benzothiazole", "indazole", "thiazinane",
        "thiomorpholine"
    ]

    exact_identifier_match = any(x in text for x in compound_terms)
    cancer_match = any(x in text for x in cancer_terms)
    scaffold_match = any(x in text for x in scaffold_terms)

    if exact_identifier_match:
        category = "Potential exact-identifier match — manual verification"
    elif cancer_match and scaffold_match:
        category = "Potential cancer/scaffold relevance — not exact compound"
    elif scaffold_match:
        category = "Scaffold-related — cancer relevance not established"
    else:
        category = "No obvious relevance from title/abstract"

    return pd.Series({
        "Cancer_Keyword_Match": cancer_match,
        "Scaffold_Keyword_Match": scaffold_match,
        "Exact_Identifier_Match": exact_identifier_match,
        "Screening_Category": category,
        "Manual_Verification": "Required"
    })

screened = pd.concat([papers, papers.apply(screen, axis=1)], axis=1)

out = R / "20_PAPSS2_Literature_Relevance_Screening.xlsx"

with pd.ExcelWriter(out, engine="openpyxl") as writer:
    screened.to_excel(writer, sheet_name="Screened_Records", index=False)

print("LITERATURE SCREENING")
print("====================")
print(screened["Screening_Category"].value_counts().to_string())
print("\nRecords requiring manual verification:", len(screened))
print("\nSaved:", out)

display(screened[[
    "PubChem_CID", "Paper_Title", "DOI",
    "Screening_Category", "Manual_Verification"
]])
```

```python
import pandas as pd
import requests
import time
from pathlib import Path
from xml.etree import ElementTree as ET

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
R = BASE / "02_Analysis_Results"

docking = pd.read_excel(R / "04_Authoritative_Docking_Results.xlsx")

session = requests.Session()
session.headers.update({"User-Agent": "PAPSS2-literature-verification/1.0"})

cancer_terms = (
    '(cancer OR carcinoma OR tumor OR tumour OR anticancer '
    'OR antitumor OR cytotoxic OR breast cancer)'
)

records = []
errors = []

for _, row in docking.iterrows():
    cid = str(row["PubChem CID"]).replace(".0", "").strip()

    candidates = []
    for col in ["PubChem Title", "IUPAC Name", "Molecule Name"]:
        value = row.get(col)
        if pd.notna(value) and str(value).strip():
            candidates.append(str(value).strip())

    # Search using distinct available compound names
    queries = list(dict.fromkeys(
        f'"{name}" AND {cancer_terms}' for name in candidates
    ))

    if not queries:
        errors.append({"PubChem_CID": cid, "Error": "No compound name available"})
        continue

    for query in queries:
        try:
            search = session.get(
                "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi",
                params={
                    "db": "pubmed",
                    "term": query,
                    "retmode": "json",
                    "retmax": 20
                },
                timeout=30
            )
            search.raise_for_status()
            ids = search.json()["esearchresult"].get("idlist", [])

            if ids:
                fetch = session.get(
                    "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esummary.fcgi",
                    params={
                        "db": "pubmed",
                        "id": ",".join(ids),
                        "retmode": "json"
                    },
                    timeout=30
                )
                fetch.raise_for_status()
                summaries = fetch.json()["result"]

                for pmid in ids:
                    item = summaries.get(pmid, {})
                    records.append({
                        "PubChem_CID": cid,
                        "Search_Query": query,
                        "PMID": pmid,
                        "Title": item.get("title", ""),
                        "Year": item.get("pubdate", ""),
                        "Journal": item.get("fulljournalname", ""),
                        "DOI": next(
                            (
                                x.get("value", "")
                                for x in item.get("articleids", [])
                                if x.get("idtype") == "doi"
                            ),
                            ""
                        ),
                        "PubMed_URL": f"https://pubmed.ncbi.nlm.nih.gov/{pmid}/"
                    })

            time.sleep(0.35)

        except Exception as e:
            errors.append({
                "PubChem_CID": cid,
                "Search_Query": query,
                "Error": str(e)
            })

columns = [
    "PubChem_CID", "Search_Query", "PMID",
    "Title", "Year", "Journal", "DOI", "PubMed_URL"
]

results = pd.DataFrame(records, columns=columns)

if not results.empty:
    results = results.drop_duplicates(subset=["PubChem_CID", "PMID"])

error_df = pd.DataFrame(errors)

output = R / "21_PAPSS2_PubMed_Exact_Compound_Verification.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    results.to_excel(writer, sheet_name="PubMed_Results", index=False)
    error_df.to_excel(writer, sheet_name="Search_Errors", index=False)

print("PUBMED VERIFICATION")
print("===================")
print("Candidates searched:", docking["PubChem CID"].nunique())
print("PubMed records retrieved:", len(results))
print("Unique PMIDs:", results["PMID"].nunique())
print("Search errors:", len(error_df))
print("\nSaved:", output)

if not results.empty:
    display(results[["PubChem_CID", "Title", "Year", "DOI"]])
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
R = BASE / "02_Analysis_Results"

df = pd.read_excel(
    R / "22_PAPSS2_PubMed_Abstract_Verification.xlsx",
    sheet_name="Abstracts"
)

def classify(row):
    title = str(row.get("Title", "")).lower()
    abstract = str(row.get("Abstract", "")).lower()
    text = title + " " + abstract

    breast = any(x in text for x in [
        "breast cancer", "breast carcinoma", "mcf-7",
        "mda-mb-231", "mcf-10a", "triple-negative breast"
    ])

    cancer = any(x in text for x in [
        "anticancer", "anti-cancer", "antitumor", "anti-tumor",
        "cytotoxic", "cancer cell", "carcinoma", "malignant"
    ])

    papss2 = "papss2" in text or "paps synthase 2" in text

    scaffold = any(x in text for x in [
        "thiophene", "benzimidazole", "benzoxazole",
        "benzothiazole", "indazole", "thiazinane",
        "thiomorpholine"
    ])

    if papss2 and breast:
        category = "PAPSS2 and breast-cancer evidence"
    elif breast and cancer:
        category = "Breast-cancer experimental/relevant evidence"
    elif papss2:
        category = "PAPSS2-related evidence; cancer relevance requires review"
    elif cancer and scaffold:
        category = "Cancer and related-scaffold evidence"
    elif cancer:
        category = "General cancer evidence; scaffold relevance unclear"
    elif scaffold:
        category = "Scaffold evidence only"
    else:
        category = "No clear relevance"

    return pd.Series({
        "Breast_Cancer_Match": breast,
        "Cancer_Match": cancer,
        "PAPSS2_Match": papss2,
        "Scaffold_Match": scaffold,
        "Abstract_Relevance": category,
        "Exact_Docked_Compound_Confirmed": "No — not established by keyword screening"
    })

screened = pd.concat([df, df.apply(classify, axis=1)], axis=1)

output = R / "23_PAPSS2_Abstract_Relevance_Classification.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    screened.to_excel(writer, sheet_name="All_Classified_Abstracts", index=False)

print("ABSTRACT-LEVEL CLASSIFICATION")
print("=============================")
print("Unique papers:", screened["PMID"].nunique())
print("\nClassification counts:")
print(screened["Abstract_Relevance"].value_counts().to_string())

print("\nKeyword matches:")
print("Breast cancer:", int(screened["Breast_Cancer_Match"].sum()))
print("General cancer:", int(screened["Cancer_Match"].sum()))
print("PAPSS2:", int(screened["PAPSS2_Match"].sum()))
print("Related scaffold:", int(screened["Scaffold_Match"].sum()))

print("\nSaved:", output)
```

```python
import pandas as pd
from pathlib import Path

BASE = Path("/content/PAPSS2_Top10_New_Derivatives")
R = BASE / "02_Analysis_Results"
FINAL = BASE / "03_Final_Results"
FINAL.mkdir(parents=True, exist_ok=True)

source = R / "17_PAPSS2_Integrated_Top10_Comparison.xlsx"
df = pd.read_excel(source, sheet_name="Integrated_Top10")

# Verify essential fields
required = [
    "PubChem CID", "Edock", "Edock_Rank_Recalculated",
    "cLogP_RDKit", "TPSA", "QED", "Total_Property_Flags",
    "Classifier_Signals_At_Or_Above_0.5",
    "Structure_Identity_Flag", "Isotope_Identity_Warning"
]
missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# One row per candidate
decision = df[required].copy()
decision["PubChem CID"] = decision["PubChem CID"].astype(str)

# Evidence-aware interpretation; no arbitrary composite score
def assess(row):
    cid = row["PubChem CID"]
    rank = int(row["Edock_Rank_Recalculated"])
    flags = int(row["Total_Property_Flags"])
    tox = int(row["Classifier_Signals_At_Or_Above_0.5"])

    if cid == "49849336":
        return pd.Series([
            "HOLD — identity discrepancy",
            "Do not advance until isotope-labelled identity is resolved."
        ])

    if cid == "58965179":
        return pd.Series([
            "Priority follow-up candidate",
            "No property flags and moderate docking rank; however, predicted hERG and skin-reaction outputs require follow-up."
        ])

    if cid == "68870353":
        return pd.Series([
            "Secondary comparator",
            "Favourable exploratory property/toxicity profile, but weaker docking rank; retain as a comparator rather than a lead."
        ])

    if rank <= 4:
        return pd.Series([
            "Low priority — liability concerns",
            "Strong docking rank, but multiple property flags and numerous elevated toxicity-model outputs."
        ])

    if flags >= 3 or tox >= 4:
        return pd.Series([
            "Low priority — predicted liabilities",
            "Multiple exploratory property flags and/or elevated toxicity-model outputs."
        ])

    return pd.Series([
        "Reserve",
        "Insufficient evidence for advancement; retain for comparison only."
    ])

decision[["Decision_Category", "Decision_Rationale"]] = decision.apply(assess, axis=1)

decision = decision.sort_values("Edock_Rank_Recalculated")

decision["Evidence_Limitations"] = (
    "No confirmed exact-compound anticancer evidence; "
    "no PAPSS2-specific PubMed abstract match; "
    "no protein-ligand interaction data; "
    "no experimental validation."
)

decision["Lead_Selection_Status"] = "No validated lead selected"

output = FINAL / "24_PAPSS2_Final_Candidate_Decision_Matrix.xlsx"

with pd.ExcelWriter(output, engine="openpyxl") as writer:
    decision.to_excel(writer, sheet_name="Final_Decision_Matrix", index=False)

    summary = pd.DataFrame({
        "Assessment": [
            "Candidates evaluated",
            "Exact-compound anticancer evidence confirmed",
            "PAPSS2-specific PubMed abstract matches",
            "Protein-ligand interaction evidence",
            "Validated lead selected",
            "Priority follow-up candidate",
            "Secondary comparator",
            "Identity discrepancy"
        ],
        "Result": [
            len(decision),
            0,
            0,
            "Unavailable",
            "No",
            "CID 58965179",
            "CID 68870353",
            "CID 49849336"
        ]
    })
    summary.to_excel(writer, sheet_name="Evidence_Summary", index=False)

print("FINAL DECISION MATRIX SAVED")
print("===========================")
print(decision[
    ["PubChem CID", "Edock_Rank_Recalculated",
     "Decision_Category", "Decision_Rationale"]
].to_string(index=False))
print("\nSaved:", output)
```

```python
# ================================================================
# PAPSS2 TOP 10 NEW DERIVATIVES — FINAL CLEAN PACKAGE
# Rebuild | Enrich compound names | Sequence files | Report | ZIP
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import shutil
import zipfile
import hashlib
import re
import os

from docx import Document
from docx.shared import Inches, Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH

SOURCE = Path("/content/PAPSS2_Top10_New_Derivatives")
FINAL = Path("/content/PAPSS2_Top10_New_Derivatives_FINAL")

if not SOURCE.exists():
    raise FileNotFoundError(f"Working project folder not found: {SOURCE}")

if FINAL.exists():
    shutil.rmtree(FINAL)

# ---------- 1. CREATE CLEAN FOLDER STRUCTURE ----------

folders = {
    "01_Source_Files": FINAL / "01_Source_Files",
    "02_Data_Quality_and_Structure_Audit": FINAL / "02_Data_Quality_and_Structure_Audit",
    "03_Docking_Results": FINAL / "03_Docking_Results",
    "04_Molecular_Properties_and_ADMET": FINAL / "04_Molecular_Properties_and_ADMET",
    "05_Literature_Analysis": FINAL / "05_Literature_Analysis",
    "06_Integrated_Candidate_Assessment": FINAL / "06_Integrated_Candidate_Assessment",
    "07_Final_Reports": FINAL / "07_Final_Reports",
    "08_Methodology": FINAL / "08_Methodology",
}

for p in folders.values():
    p.mkdir(parents=True, exist_ok=True)

# ---------- 2. BUILD AUTHORITATIVE CID-TO-NAME MAP ----------

input_xlsx = (
    SOURCE / "01_Input_Files" /
    "PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx"
)
input_sdf = (
    SOURCE / "01_Input_Files" /
    "PAPSS2_TOP10_COMPOUNDS.sdf"
)

if not input_xlsx.exists() or not input_sdf.exists():
    raise FileNotFoundError("One or both original input files are missing.")

raw = pd.read_excel(input_xlsx)

required = ["PubChem CID", "Edock"]
for c in required:
    if c not in raw.columns:
        raise ValueError(f"Authoritative workbook missing column: {c}")

def clean_cid(x):
    if pd.isna(x):
        return None
    s = str(x).strip()
    if s.endswith(".0"):
        s = s[:-2]
    return s

raw["PubChem CID"] = raw["PubChem CID"].apply(clean_cid)

def first_valid(row, columns):
    for col in columns:
        if col in row.index and pd.notna(row[col]):
            value = str(row[col]).strip()
            if value and value.lower() not in ["nan", "none"]:
                return value
    return "Name unavailable in source workbook"

name_columns = ["PubChem Title", "IUPAC Name", "Molecule Name", "NAME"]

name_map = {}
for _, row in raw.iterrows():
    cid = row["PubChem CID"]
    if cid:
        name_map[cid] = first_valid(row, name_columns)

if len(name_map) != 10:
    raise ValueError(f"Expected 10 unique candidate CIDs; found {len(name_map)}")

# Prefer IUPAC name when PubChem Title is absent; retain full source names.
candidate_master = raw.copy()
candidate_master["Compound Name"] = candidate_master.apply(
    lambda r: first_valid(r, name_columns), axis=1
)

candidate_master["Edock_Rank_Recalculated"] = (
    candidate_master["Edock"].rank(method="min", ascending=True).astype(int)
)

candidate_master = candidate_master.sort_values(
    "Edock_Rank_Recalculated"
).reset_index(drop=True)

# ---------- 3. COPY INPUT FILES WITH CLEAR NAMES ----------

shutil.copy2(
    input_xlsx,
    folders["01_Source_Files"] /
    "01_Authoritative_PAPSS2_Top10_Docking_Input.xlsx"
)
shutil.copy2(
    input_sdf,
    folders["01_Source_Files"] /
    "02_PAPSS2_Top10_Ligand_Structures.sdf"
)

# ---------- 4. CLASSIFY AND COPY ALL EXISTING ANALYSIS WORKBOOKS ----------

analysis_dir = SOURCE / "02_Analysis_Results"

if not analysis_dir.exists():
    raise FileNotFoundError("Analysis results folder is missing.")

file_routes = [
    ("audit", folders["02_Data_Quality_and_Structure_Audit"]),
    ("structure", folders["02_Data_Quality_and_Structure_Audit"]),
    ("cid", folders["02_Data_Quality_and_Structure_Audit"]),
    ("docking", folders["03_Docking_Results"]),
    ("property", folders["04_Molecular_Properties_and_ADMET"]),
    ("admet", folders["04_Molecular_Properties_and_ADMET"]),
    ("toxicity", folders["04_Molecular_Properties_and_ADMET"]),
    ("literature", folders["05_Literature_Analysis"]),
    ("crossref", folders["05_Literature_Analysis"]),
    ("pubmed", folders["05_Literature_Analysis"]),
    ("abstract", folders["05_Literature_Analysis"]),
    ("integrated", folders["06_Integrated_Candidate_Assessment"]),
    ("priority", folders["06_Integrated_Candidate_Assessment"]),
]

def destination_for(filename):
    low = filename.lower()
    for keyword, folder in file_routes:
        if keyword in low:
            return folder
    return folders["06_Integrated_Candidate_Assessment"]

def safe_sheet_name(name):
    return str(name)[:31]

def enrich_workbook(src, dst):
    xls = pd.ExcelFile(src)
    with pd.ExcelWriter(dst, engine="openpyxl") as writer:
        for sheet in xls.sheet_names:
            data = pd.read_excel(src, sheet_name=sheet)

            # Add Compound Name beside a single-CID column.
            cid_cols = [
                c for c in data.columns
                if str(c).strip().lower() in [
                    "pubchem cid", "cid", "candidate cid"
                ]
            ]

            for cid_col in cid_cols:
                name_col = "Compound Name"
                if name_col in data.columns:
                    name_col = "Compound Name (matched to CID)"

                names = data[cid_col].apply(
                    lambda x: name_map.get(clean_cid(x), "Name unavailable")
                )
                insert_at = data.columns.get_loc(cid_col) + 1
                data.insert(insert_at, name_col, names)

            # Add names for semicolon-separated candidate CID lists.
            for cid_col in [
                c for c in data.columns
                if "associated_candidate_cids" in str(c).lower()
            ]:
                name_col = "Associated_Candidate_Names"
                if name_col not in data.columns:
                    def map_multiple(x):
                        if pd.isna(x):
                            return ""
                        ids = [clean_cid(v) for v in str(x).split(";")]
                        return "; ".join(
                            f"{cid}: {name_map.get(cid, 'Name unavailable')}"
                            for cid in ids if cid
                        )
                    data[name_col] = data[cid_col].apply(map_multiple)

            data.to_excel(
                writer,
                sheet_name=safe_sheet_name(sheet),
                index=False
            )

# Human-readable sequence numbers within each section.
section_counters = {key: 0 for key in folders}
copied_analysis = []

for src in sorted(analysis_dir.glob("*.xlsx")):
    target_folder = destination_for(src.name)
    section_key = next(k for k, v in folders.items() if v == target_folder)
    section_counters[section_key] += 1
    seq = section_counters[section_key]

    stem = re.sub(r"^\d+_", "", src.stem)
    target = target_folder / f"{seq:02d}_{stem}_NAMED.xlsx"

    enrich_workbook(src, target)
    copied_analysis.append(target)

# ---------- 5. FINAL NAMED CANDIDATE MASTER TABLE ----------

decision_file = (
    SOURCE / "03_Final_Results" /
    "24_PAPSS2_Final_Candidate_Decision_Matrix.xlsx"
)

if not decision_file.exists():
    raise FileNotFoundError("Final decision matrix is missing.")

decision = pd.read_excel(
    decision_file,
    sheet_name="Final_Decision_Matrix"
)

decision["PubChem CID"] = decision["PubChem CID"].apply(clean_cid)
decision.insert(
    1,
    "Compound Name",
    decision["PubChem CID"].map(name_map).fillna("Name unavailable")
)

# Ensure full names are not truncated in the workbook data.
decision = decision.sort_values("Edock_Rank_Recalculated")

# Add authoritative docking and molecular data from integrated workbook.
integrated_path = (
    SOURCE / "02_Analysis_Results" /
    "17_PAPSS2_Integrated_Top10_Comparison.xlsx"
)
integrated = pd.read_excel(integrated_path, sheet_name="Integrated_Top10")
integrated["PubChem CID"] = integrated["PubChem CID"].apply(clean_cid)
integrated.insert(
    1,
    "Compound Name",
    integrated["PubChem CID"].map(name_map).fillna("Name unavailable")
)

# Final workbook: named master comparison, decision, evidence audit.
final_master = integrated.sort_values("Edock_Rank_Recalculated")

final_summary_path = (
    folders["07_Final_Reports"] /
    "01_PAPSS2_Final_Named_Candidate_Comparison.xlsx"
)

with pd.ExcelWriter(final_summary_path, engine="openpyxl") as writer:
    final_master.to_excel(writer, sheet_name="Named_Top10_Comparison", index=False)
    decision.to_excel(writer, sheet_name="Named_Final_Decisions", index=False)

    candidate_master.to_excel(
        writer, sheet_name="Authoritative_Docking_Input", index=False
    )

    pd.DataFrame({
        "Assessment": [
            "Number of candidates",
            "PubMed abstracts screened",
            "Confirmed exact-compound anticancer evidence",
            "PAPSS2-specific PubMed abstract matches",
            "Protein-ligand interaction data",
            "Validated lead selected",
            "Priority follow-up candidate",
            "Secondary comparator",
            "Identity discrepancy"
        ],
        "Result": [
            10, 47, 0, 0, "Unavailable", "No",
            "CID 58965179",
            "CID 68870353",
            "CID 49849336"
        ]
    }).to_excel(writer, sheet_name="Evidence_Summary", index=False)

# ---------- 6. FINAL DECISION MATRIX WITH FULL NAMES ----------

decision_path = (
    folders["07_Final_Reports"] /
    "02_PAPSS2_Final_Decision_Matrix_Named.xlsx"
)

with pd.ExcelWriter(decision_path, engine="openpyxl") as writer:
    decision.to_excel(writer, sheet_name="Candidate_Decisions", index=False)

# ---------- 7. SUPERVISOR-READY DOCX REPORT ----------

doc = Document()
doc.sections[0].top_margin = Inches(0.65)
doc.sections[0].bottom_margin = Inches(0.65)
doc.styles["Normal"].font.name = "Arial"
doc.styles["Normal"].font.size = Pt(9)

heading = doc.add_heading(
    "PAPSS2 Top-10 Docking Candidate Evaluation",
    0
)
heading.alignment = WD_ALIGN_PARAGRAPH.CENTER

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
p.add_run(
    "Integrated computational assessment of docking, molecular properties, "
    "ADMET predictions and literature evidence"
).italic = True

doc.add_heading("1. Objective", level=1)
doc.add_paragraph(
    "To evaluate ten PAPSS2 docking candidates by integrating the supplied "
    "ICM-Pro docking scores, ligand structure checks, RDKit molecular descriptors, "
    "ADMET-AI predictions and literature screening. The assessment is intended "
    "for computational prioritization and does not establish biological activity."
)

doc.add_heading("2. Input data and provenance", level=1)
doc.add_paragraph(
    "The authoritative docking data source was "
    "PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx. Ligand structures were supplied "
    "in PAPSS2_TOP10_COMPOUNDS.sdf. The workbook's target PDB field is recorded "
    "as 8I10 (zero); this identifier has not been independently resolved."
)

doc.add_heading("3. Candidate comparison", level=1)
doc.add_paragraph(
    "The table reports the compound name alongside the PubChem CID to ensure "
    "unambiguous candidate identification."
)

table = doc.add_table(rows=1, cols=5)
table.style = "Light Shading Accent 1"
for cell, value in zip(
    table.rows[0].cells,
    ["Edock rank", "PubChem CID", "Compound name", "Edock score", "Decision"]
):
    cell.text = value

decision_lookup = {
    str(row["PubChem CID"]): row["Decision_Category"]
    for _, row in decision.iterrows()
}

for _, row in final_master.iterrows():
    cid = clean_cid(row["PubChem CID"])
    cells = table.add_row().cells
    cells[0].text = str(row["Edock_Rank_Recalculated"])
    cells[1].text = cid
    cells[2].text = str(row["Compound Name"])
    cells[3].text = str(row["Edock"])
    cells[4].text = decision_lookup.get(cid, "Review required")

doc.add_heading("4. Integrated interpretation", level=1)
doc.add_paragraph(
    "The strongest docking scores were observed for the four polysulfanyl-thiophene "
    "candidates. However, these candidates also showed multiple exploratory "
    "molecular-property flags and several elevated ADMET-AI classifier outputs. "
    "Docking rank alone was therefore not used as the selection criterion."
)
doc.add_paragraph(
    "CID 58965179 was retained as the priority follow-up candidate because it "
    "had no flags in the recorded molecular-property assessment and a moderate "
    "docking rank. Its predicted hERG and skin-reaction outputs remain concerns. "
    "CID 68870353 was retained as a secondary comparator because of its more "
    "favourable exploratory property and toxicity-prediction profile, despite "
    "its weaker docking rank."
)
doc.add_paragraph(
    "CID 49849336 remains on hold because the workbook identifies an isotope-labelled "
    "compound while the corresponding SDF structure is unlabelled. Predictions "
    "calculated from the SDF must not be assigned to the isotope-labelled identity."
)

doc.add_heading("5. Literature assessment", level=1)
doc.add_paragraph(
    "Forty-seven PubMed abstracts were retrieved and screened. Automated "
    "classification identified three breast-cancer keyword matches, sixteen "
    "related-scaffold keyword matches and no PAPSS2 keyword matches. These "
    "categories are screening indicators, not confirmation of exact-compound "
    "anticancer activity. No exact-compound anticancer evidence was established."
)

doc.add_heading("6. Final conclusion", level=1)
doc.add_paragraph(
    "No validated lead is selected. CID 58965179 is the priority follow-up "
    "candidate and CID 68870353 is a secondary comparator. Both are provisional "
    "computational prioritizations requiring further structural and experimental "
    "evaluation."
)

doc.add_heading("7. Limitations", level=1)
limitations = [
    "No docked protein–ligand complex files were supplied; protein–ligand interactions, binding residues and pose geometry could not be assessed.",
    "Receptor preparation, binding-site definition, docking parameters, software version and redocking validation were not fully documented.",
    "Docking scores are computational estimates and may be affected by ligand size, lipophilicity and scoring-function limitations.",
    "RDKit descriptors and ADMET-AI outputs are predictions, not experimental measurements.",
    "The 0.5 classifier threshold is exploratory and is not a universal toxicity cutoff.",
    "The LD50_Zhu endpoint was not used to rank safety because its scale and direction were not independently established here.",
    "Literature keyword matches require manual full-text and chemical-structure verification.",
    "No experimental PAPSS2 inhibition, breast-cancer cell activity, selectivity or in vivo validation was available."
]
for item in limitations:
    doc.add_paragraph(item, style="List Bullet")

doc.add_heading("8. Recommended next steps", level=1)
for item in [
    "Resolve the target structure identifier and verify the original ICM-Pro receptor setup.",
    "Resolve the isotope identity discrepancy for CID 49849336.",
    "Retrieve the original docked poses and inspect protein–ligand interactions.",
    "Validate the docking protocol using a suitable reference ligand and pose comparison.",
    "Manually verify exact-compound literature using chemical structures and synonyms.",
    "Perform experimental PAPSS2 inhibition and relevant selectivity/cytotoxicity assays before biological claims."
]:
    doc.add_paragraph(item, style="List Number")

report_path = (
    folders["07_Final_Reports"] /
    "03_PAPSS2_Supervisor_Ready_Evaluation_Report.docx"
)
doc.save(report_path)

# ---------- 8. METHODOLOGY DOCUMENT ----------

method = Document()
method.styles["Normal"].font.name = "Arial"
method.styles["Normal"].font.size = Pt(10)
method.add_heading("PAPSS2 Computational Evaluation — Methodology", 0)

method_sections = {
    "1. Source files": [
        "Authoritative docking workbook: PAPSS2_FINAL_TOP10_COMPOUNDS_docked.xlsx.",
        "Ligand structure file: PAPSS2_TOP10_COMPOUNDS.sdf.",
        "The Excel workbook was treated as the authoritative source for Edock scores."
    ],
    "2. Docking ranking": [
        "Edock scores were ranked in ascending order; more negative values received higher rank.",
        "The original Final Rank field was retained separately from the recalculated Edock rank.",
        "The workbook records the target PDB as 8I10 (zero); independent verification remains necessary."
    ],
    "3. Structure checks": [
        "Ten SDF records were assessed for parsing and sanitization.",
        "Structures were matched to the workbook PubChem CIDs.",
        "CID 49849336 has an isotope-labelled identity in the workbook but an unlabelled SDF structure."
    ],
    "4. Molecular descriptors": [
        "RDKit descriptors included molecular weight, calculated logP, TPSA, QED and Lipinski violations.",
        "Property flags are exploratory indicators and are not experimental measurements."
    ],
    "5. ADMET": [
        "ADMET-AI predictions were generated for the ten SDF structures.",
        "Toxicity classifier outputs were reviewed as model predictions.",
        "The 0.5 threshold is exploratory, not a universal toxicity threshold."
    ],
    "6. Literature": [
        "Literature records were assembled through OpenAlex, Crossref and PubMed.",
        "Forty-seven unique PubMed abstracts were screened using keyword indicators.",
        "Keyword matching does not establish exact-compound activity or PAPSS2 inhibition."
    ],
    "7. Integrated assessment": [
        "Docking, molecular properties, ADMET predictions and literature were considered together.",
        "No arbitrary weighted composite score was applied.",
        "CID 58965179 is the priority follow-up candidate; CID 68870353 is a secondary comparator.",
        "No validated lead was selected."
    ],
    "8. Limitations": [
        "Protein–ligand complex files were unavailable, so interaction analysis could not be performed.",
        "Full receptor preparation and docking settings were not available for independent reproduction.",
        "No experimental target inhibition or anticancer validation was available."
    ]
}

for h, points in method_sections.items():
    method.add_heading(h, level=1)
    for point in points:
        method.add_paragraph(point, style="List Bullet")

method_path = (
    folders["08_Methodology"] /
    "01_PAPSS2_Computational_Methodology_and_Limitations.docx"
)
method.save(method_path)

# ---------- 9. COPY AUDIT / EVIDENCE FILES TO FINAL SECTIONS ----------

# Save candidate name mapping as a standalone reference.
name_reference = pd.DataFrame({
    "PubChem CID": list(name_map.keys()),
    "Compound Name": list(name_map.values())
})

name_reference_path = (
    folders["01_Source_Files"] /
    "03_PAPSS2_Candidate_CID_Name_Reference.xlsx"
)
name_reference.to_excel(name_reference_path, index=False)

# ---------- 10. CONTENTS INDEX ----------

manifest_rows = []
for path in sorted(FINAL.rglob("*")):
    if path.is_file():
        manifest_rows.append({
            "Relative_Path": str(path.relative_to(FINAL)),
            "File_Name": path.name,
            "File_Type": path.suffix.lower(),
            "Size_KB": round(path.stat().st_size / 1024, 2)
        })

manifest = pd.DataFrame(manifest_rows)
manifest_path = FINAL / "00_README_Project_File_Index.xlsx"
manifest.to_excel(manifest_path, index=False)

readme = FINAL / "00_README_Project_File_Index.txt"
with open(readme, "w", encoding="utf-8") as f:
    f.write("PAPSS2 TOP 10 NEW DERIVATIVES — FINAL PROJECT PACKAGE\n")
    f.write("=" * 65 + "\n\n")
    f.write("Folder sequence:\n")
    f.write("01_Source_Files — original input files and candidate name reference\n")
    f.write("02_Data_Quality_and_Structure_Audit — source and structure audits\n")
    f.write("03_Docking_Results — authoritative docking results\n")
    f.write("04_Molecular_Properties_and_ADMET — RDKit and ADMET outputs\n")
    f.write("05_Literature_Analysis — literature retrieval and screening\n")
    f.write("06_Integrated_Candidate_Assessment — integrated evidence and priority workbooks\n")
    f.write("07_Final_Reports — named final tables, decision matrix and supervisor report\n")
    f.write("08_Methodology — methodology and limitations\n\n")
    f.write("Compound identification: candidate tables include PubChem CID and compound name.\n")
    f.write("Conclusion: no validated lead selected.\n")
    f.write("Priority follow-up: CID 58965179.\n")
    f.write("Secondary comparator: CID 68870353.\n")
    f.write("CID 49849336: hold pending isotope identity resolution.\n\n")
    f.write("All model predictions are computational and require experimental validation.\n")

# ---------- 11. FINAL AUDIT ----------

all_files = [p for p in FINAL.rglob("*") if p.is_file()]
empty_files = [str(p.relative_to(FINAL)) for p in all_files if p.stat().st_size == 0]
empty_folders = [
    str(p.relative_to(FINAL))
    for p in FINAL.rglob("*")
    if p.is_dir() and not any(p.iterdir())
]

hashes = {}
duplicates = []
for p in all_files:
    h = hashlib.sha256(p.read_bytes()).hexdigest()
    if h in hashes:
        duplicates.append((hashes[h], str(p.relative_to(FINAL))))
    else:
        hashes[h] = str(p.relative_to(FINAL))

if empty_files or empty_folders or duplicates:
    print("Audit warnings:")
    print("Empty files:", empty_files)
    print("Empty folders:", empty_folders)
    print("Exact duplicates:", duplicates)

# ---------- 12. ZIP AND INTEGRITY CHECK ----------

zip_path = Path("/content/PAPSS2_Top10_New_Derivatives_FINAL.zip")
if zip_path.exists():
    zip_path.unlink()

shutil.make_archive(
    str(zip_path).removesuffix(".zip"),
    "zip",
    root_dir=FINAL.parent,
    base_dir=FINAL.name
)

with zipfile.ZipFile(zip_path, "r") as z:
    bad = z.testzip()
    zip_members = z.namelist()

if bad:
    raise RuntimeError(f"ZIP integrity check failed: {bad}")

# ---------- 13. FINAL SUMMARY ----------

print("\n" + "=" * 70)
print("PAPSS2 FINAL PACKAGE — VERIFIED")
print("=" * 70)
print("Project folder:", FINAL)
print("ZIP archive:", zip_path)
print("Candidate count:", len(name_map))
print("Named final comparison:", final_summary_path.exists())
print("Named decision matrix:", decision_path.exists())
print("Supervisor report:", report_path.exists())
print("Methodology document:", method_path.exists())
print("Files in final folder:", len(all_files))
print("Files in ZIP:", len(zip_members))
print("Empty files:", len(empty_files))
print("Empty folders:", len(empty_folders))
print("Exact duplicate files:", len(duplicates))
print("ZIP integrity:", "PASS" if bad is None else "FAIL")
print("\nFinal conclusion: NO VALIDATED LEAD SELECTED")
print("Priority follow-up: CID 58965179")
print("Secondary comparator: CID 68870353")
print("Identity hold: CID 49849336")
print("\nDownloading final ZIP...")

from google.colab import files
files.download(str(zip_path))
```
